## Accessing Dataset Files

To access datasets (usually files), you can either:

- set a path variable to the data folder, which will be easier in the long run; or
- write path (full or relative) and file names as arguments

You are learning these basic operations now and will probably just use **relative path** in the future.


When importing datasets, some path information are good to know:

1. **Full paths**:

   - **macOS** (and Linux): A full/absolute path starts with "`/`" (the "root" directory) or you can use "`~`" to start with the user's home directory (e.g., data_path = "/Users/<user>`/`workspace/data/" == "`~`/workspace/data/").

   - **Windows**: A full path includes a drive letter and backslashes (`\`) for path separator (e.g. data_path = "`C:`\Users\<user>\workspace\dsm\data\") though forward slashes usually work as well.

2. **Relative paths**: Usually it's just easier to use a relative path to access the data folders (e.g., `../../data/`).

3. Current working directory (CWD/PWD): To know your Current/Present Working Directory (this notebook's location), use `os.getcwd()` or `%pwd`.
   
4. Home directory: To write portable code across OSs, we prefer `os.path.expanduser`, which can resolve the directory passed (e.g., `~`) for us to use.

5. Join paths: This is used to join the path variable with the paths/filenames instead of hardcoding separators and literals. For example, `os.path.join( )` to join a path and a dataset file.


As an example, the `os.getcwd()` function will give us the full path of the notebook you are working on, which can be critical when using relative paths.


`````python
import os   ### os is a standard library module for interacting with the OS.
CWD = os.getcwd()
CWD  # print current working directory
`````

## Reading Dataset Files


Now let us read the dataset file using `pd.read_csv( )`. You may read the data file directly or and open it as a file and save it a variable so you can manipulate the data file.


`````python
# now we can read the data file using pandas
# this is one way to do it
home = os.path.expanduser('~')
data_path = os.path.join(home, 'workspace', 'dsm', 'data')
data = os.path.join(data_path, 'data.csv')
with open(data) as f:  # open the file for reading and assign the file object variable f
    df = pd.read_csv(f)  # read the csv file into a DataFrame
df.head()  # display the first 5 rows of the DataFrame
`````

`````python
# just to show the path variables we used
data_path
`````

In summary, there are different ways of reading your datasets:

#### Use relative path


`````python
# another way to do it is to use relative path
# ../../data/data.csv
data = ("../../data/data.csv")
df = pd.read_csv(data)
`````

#### Path variable & filename


`````python
df = pd.read_csv(data_path + "/data.csv")
`````

#### Use `path.join`


`````python
data = os.path.join(data_path, "data.csv")
df = pd.read_csv(data)
`````

#### Use with statement
When you need to manipulate the file.


`````python
with open(data_path+"/data.csv") as f:
    df = pd.read_csv(f)
`````

#### change dir
Alternatively, change directory into the data folder and read the data from there.

`````python
os.chdir(data_path)
df = pd.read_csv("data.csv")
df.describe()
`````
