```python
from numpy import random
```

```python
logi = random.logistic(loc=1, scale=2, size=(2, 3))
print(logi)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
random.seed(42)
logi = random.logistic(size=1000)
sns.displot(logi, kind='kde', color='blue')
# plt.show()
```

Now, do a `histplot`

```python

random.seed(42)
logi = random.logistic(size=1000)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))

sns.histplot(data=logi, color='blue', ax=axes[0])
sns.histplot(data=logi, color='green', kde=True, ax=axes[1])
sns.kdeplot(data=logi, color='red', ax=axes[2])
# sns.histplot(data=logi, color='blue', kde=True,  fill=False )
```

```python
norm = random.normal(scale=2, size=1000)
sns.distplot(norm, hist=False, color="red")
```
