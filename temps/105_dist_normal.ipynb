```python
from numpy import random
random.seed(42)
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
##### normal distribution
### the normal method in random

random.seed(42)

x = random.normal(size=(2, 3))
print(x)
```

```python
### generate a random normal distribution of size 2X3 with mean at 1 and SD of 2

random.seed(42)
x = random.normal(loc=1, scale=2, size=(2, 3))
print(x)
```

```python
random.seed(42)

np.random.seed(42)
sns.displot(random.normal(size=1000))
```

```python
random.seed(42)

sns.distplot(random.normal(size=1000), hist=False )
```

```python
##### visualization of Normal Distribution

import matplotlib as mpl
mpl.rcParams["figure.dpi"] = 100   # default ~ 100–150; lower = smaller on screen

random.seed(42)

# sns.distplot(random.normal(size=1000), hist=False)
sns.displot(random.normal(size=1000), kind="kde", height=3, aspect=5/3)
# plt.show()
```
