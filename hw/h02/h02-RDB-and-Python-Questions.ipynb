- Read the list of words using a **for loop** to count the number of the times the word "word" appears in the book. For the counter, call it word_count.

```python
##### answer here
file = '../../data/little_women.txt'
book = open(file, 'r')
string_words = book.read()
list_words = string_words.split()
word_count=0
for word in list_words:
    if word == 'the':
        # print('word')
        word_count += 1
print(word_count)
```
