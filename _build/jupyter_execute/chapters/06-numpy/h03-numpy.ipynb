### 1. Review list operations

```python
### Task: create a list of elements 0 to 9:
### 1. create an empty list variable, call it "lst"
### 2. use a for loop
### 3. use the range function
### 4. use the append() method

lst = []
for i in range(10):
    lst.append(i)
```

```python
### evaluate the list
lst
```

```python
### check the data type of the list
type(lst)
```

```python
### append() 6 to list "lst"
lst_append = 10
lst.append(lst_append)
```

```python
### extend lst using extend() to add 11, 12, 13 to lst

lst.extend([11,12,13])
```

```python
### evaluate lst
lst
```

```python
### use list **slicing** to create a new list from lst 
### still call it lst 
### of elements 0 through 9
### (reassign the list to variable name lst)

lst = lst[0:10]
```

```python
### evaluate lst
lst
```

```python
### write a for loop to remove some elements
### use the range function
### use the pop() method to remove 6, 7, 8, 9 from the list
### *note: pop() removes the last element by default if you don't assign index

for i in range(4):
    lst.pop()
```

```python
### evaluate lst
lst
```

```python
### Use the numpy array function (if it's a method, you have to use it with an object)
### turn the list lst into a numpy array
### assign the numpy array to a variable, call it "arr"

arr = np.array(lst)
```
