## NumPy

```python
import numpy as np
arr_2d = np.array([
    [5, 10, 15],
    [20, 25, 30],
    [35, 40, 45]
])

print("Element at [1, 2]:", arr_2d[0, 1])
print("Second row:", arr_2d[1])
print("First column:", arr_2d[:1])
print("Bottom-right 2x2 subarray:\n", arr_2d[1, 1:])
```
