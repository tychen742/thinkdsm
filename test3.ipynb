| Plot Type            | pandas                                  | Matplotlib                                      | Seaborn                                                        | Best For                          |
| -------------------- | --------------------------------------- | ----------------------------------------------- | -------------------------------------------------------------- | --------------------------------- |
| **Line**             | `.plot.line()` / `.plot()`              | `plt.plot()`                                    | `sns.lineplot()`                                               | Time series, trends               |
| **Scatter**          | `.plot.scatter(x, y)`                   | `plt.scatter()`                                 | `sns.scatterplot()`                                            | Correlations, relationships       |
| **Bar (vertical)**   | `.plot.bar()`                           | `plt.bar()`                                     | `sns.barplot()`                                                | Category comparisons              |
| **Bar (horizontal)** | `.plot.barh()`                          | `plt.barh()`                                    | `sns.barplot(y=..., x=...)`                                    | Long category labels              |
| **Histogram**        | `.plot.hist()`                          | `plt.hist()`                                    | `sns.histplot()`                                               | Distribution of numeric data      |
| **Box plot**         | `.plot.box()`                           | `plt.boxplot()`                                 | `sns.boxplot()`                                                | Distribution + outliers           |
| **KDE/Density**      | `.plot.kde()` / `.plot.density()`       | — *(use `scipy.stats.gaussian_kde` or Seaborn)* | `sns.kdeplot()`                                                | Smoothed distribution             |
| **ECDF**             | —                                       | — *(manual)*                                    | `sns.ecdfplot()`                                               | Cumulative distribution           |
| **Area**             | `.plot.area()`                          | `plt.fill_between()`                            | —                                                              | Cumulative trends                 |
| **Stacked area**     | `.plot.area(stacked=True)`              | `plt.stackplot()`                               | —                                                              | Composition over time             |
| **Pie**              | `.plot.pie()` *(Series only)*           | `plt.pie()`                                     | —                                                              | Part-to-whole (use sparingly)     |
| **Hexbin**           | `.plot.hexbin(x, y)` *(DataFrame only)* | `plt.hexbin()`                                  | `sns.jointplot(kind="hex")` *(figure-level)*                   | Dense scatter (binning)           |
| **Violin**           | —                                       | `plt.violinplot()`                              | `sns.violinplot()`                                             | Distribution shape by category    |
| **Heatmap**          | —                                       | `plt.imshow()` / `plt.pcolormesh()`             | `sns.heatmap()`                                                | Correlation matrices, grids       |
| **Count plot**       | `.value_counts().plot.bar()`            | — *(use `plt.bar` on counts)*                   | `sns.countplot()`                                              | Frequency of categories           |
| **Pair plot**        | —                                       | —                                               | `sns.pairplot()` *(figure-level)*                              | Multivariate relationships        |
| **Joint plot**       | —                                       | —                                               | `sns.jointplot()` *(figure-level)*                             | Bivariate + marginals             |
| **Regression**       | —                                       | `plt.plot()` + fitted curve                     | `sns.regplot()` (axes) / `sns.lmplot()` *(figure-level)*       | Linear relationships + CI         |
| **Error bars**       | —                                       | `plt.errorbar()`                                | `sns.pointplot()` / `sns.barplot()` / `sns.lineplot()` *(CIs)* | Data with uncertainty             |
| **Contour**          | —                                       | `plt.contour()` / `plt.contourf()`              | `sns.kdeplot(x, y)` *(2D density contours)*                    | Continuous 2D fields / level sets |
