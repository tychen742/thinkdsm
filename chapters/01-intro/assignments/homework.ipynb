## Question 8: Calculate Order Values

A small online store sold 36 notebooks at 3.25 dollars each and gave a 12 dollar coupon on the order. Write assignment statements that calculate `gross_sales` and `net_sales`, then print both values rounded to two decimal places.

````python
units_sold = 36
unit_price = 3.25
coupon = 12

### Your code starts here.



### Your code ends here.
````

````python
### Solution
units_sold = 36
unit_price = 3.25
coupon = 12

gross_sales = units_sold * unit_price
net_sales = gross_sales - coupon

print("Gross sales:", round(gross_sales, 2))
print("Net sales:", round(net_sales, 2))
````
