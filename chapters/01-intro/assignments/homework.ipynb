## Question 8: Expressions and Statements

A coffee shop sold 42 drinks at an average price of 4.75 dollars. It spent 96 dollars on supplies. Write assignment statements that calculate `revenue` and `profit`, then print both values rounded to two decimal places.

````python
drinks_sold = 42
average_price = 4.75
supply_cost = 96

### Your code starts here.



### Your code ends here.
````

````python
### Solution
drinks_sold = 42
average_price = 4.75
supply_cost = 96

revenue = drinks_sold * average_price
profit = revenue - supply_cost

print("Revenue:", round(revenue, 2))
print("Profit:", round(profit, 2))
````
