```python
### Lab: Customer Feedback Categories
# A manager receives the following customer feedback counts for one week.
#
# positive = 42
# neutral = 18
# negative = 10
#
# 1. Store the three counts in variables.
# 2. Calculate the total number of feedback responses.
# 3. Calculate the percentage of responses that were negative.
# 4. Print one management interpretation of the result.
### Your code starts here.



### Your code ends here.
```

```python
### Solution
positive = 42
neutral = 18
negative = 10

total = positive + neutral + negative
negative_percent = negative / total * 100

print("Total responses:", total)
print("Negative percentage:", round(negative_percent, 1))
print("Interpretation: Negative feedback is a minority of responses, but it is large enough to review common complaint themes.")
```
