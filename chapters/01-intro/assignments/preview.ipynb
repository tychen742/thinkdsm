<script>
(function () {
  function initPreviewQuiz() {
    var quiz = document.getElementById('ch01-preview-quiz');
    if (!quiz || quiz.dataset.ready === 'true') return;
    quiz.dataset.ready = 'true';

    var questions = Array.prototype.slice.call(quiz.querySelectorAll('.ch01-preview-question'));
    var result = document.getElementById('ch01-preview-result');
    var check = document.getElementById('ch01-preview-check');
    var reset = document.getElementById('ch01-preview-reset');

    check.addEventListener('click', function () {
      var correct = 0;
      var answered = 0;

      questions.forEach(function (question) {
        var selected = question.querySelector('input[type=radio]:checked');
        var feedback = question.querySelector('.ch01-preview-feedback');
        var answer = question.dataset.answer;
        feedback.className = 'ch01-preview-feedback';

        if (!selected) {
          feedback.textContent = 'Choose an answer.';
          feedback.classList.add('incorrect');
          return;
        }

        answered += 1;
        if (selected.value === answer) {
          correct += 1;
          feedback.textContent = 'Correct.';
          feedback.classList.add('correct');
        } else {
          feedback.textContent = 'Not quite. The correct answer is ' + answer + '.';
          feedback.classList.add('incorrect');
        }
      });

      result.textContent = 'Score: ' + correct + ' / ' + questions.length + ' (' + answered + ' answered)';
    });

    reset.addEventListener('click', function () {
      questions.forEach(function (question) {
        question.querySelectorAll('input[type=radio]').forEach(function (input) {
          input.checked = false;
        });
        var feedback = question.querySelector('.ch01-preview-feedback');
        feedback.className = 'ch01-preview-feedback';
        feedback.textContent = '';
      });
      result.textContent = '';
    });
  }

  if (document.readyState === 'loading') {
    document.addEventListener('DOMContentLoaded', initPreviewQuiz);
  } else {
    initPreviewQuiz();
  }
})();
</script>
