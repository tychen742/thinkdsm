# Seaborn

Seaborn is a high-level statistical visualization library built on Matplotlib. This chapter shows how Seaborn works with pandas DataFrames, how semantic mappings such as `hue`, `style`, and `size` support group comparisons, and how figure-level functions create faceted and multivariate views.

## Learning Goals

By the end of this chapter, you should be able to:

1. Create common Seaborn plots from pandas DataFrames.
2. Use tidy data and semantic mappings for grouped visualizations.
3. Compare distribution, relational, and categorical plot families.
4. Use figure-level wrappers and faceting for small-multiple comparisons.
5. Apply themes and styling choices that improve chart readability.
