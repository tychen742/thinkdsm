# Terminal & Shell

Having some knowledge of using the command line interface (CLI) is necessary when learning how to code. A CLI is often referred to as the command prompt in Windows, the “shell” in Linux/macOS, or the command line/terminal in general. 

Major operating systems (OSs) all have CLIs for users to interact with the operating system by issuing text commands in the terminal rather than using the graphical user interface (GUI). Different OSs are shipped with different default **terminal** applications (or “terminal emulators” as they emulate the locally-attached dumb terminals) to run different **shell** languages to communicate with the OSs. 

A terminal application provides access to a shell, which is a command-line/shell language interpreter that reads and executes commands entered by the user. While a terminal can support multiple shells, each shell performs the same core function of interacting with the operating system. Major OSs and their defaut shells are:

- Windows: Terminal (PowerShell, Command Prompt)
- macOS: Terminal.app (Zsh, Bash)
- Linux: the shell (usually Bash)

When you open the CLI application, you mainly use the keyboard, rather than a mouse, to issue text commands. In the terminal, you see a command prompt (e.g., `user_name@computer_name:~$` for macOS/Linux or `C:\Users\user_name>` for Windows), followed by a cursor (may blink), which indicates where you type to issue your commands for execution.

Unix-like OSs (such as Unix, macOS, and Linux) systems are file-based, meaning the design principle dictates that everything in the system is a file, and the files are organized as a tree-like file system structure. In the command line, a file path is then used to specify the location of a file in the computer’s file system structure. The file system structure begins with the root (`/`), which contains several default first-level directories representing various functionalities. For example, one of the first default locations after logging in is the user’s home directory, which is specified as `/home/USER_NAME`.
