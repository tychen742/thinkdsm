```python
from numpy import random
```

```python
expo = random.exponential(scale=2, size=(2, 3))
print(expo)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

random.seed(42)
expo = random.exponential(size=1000)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))

sns.histplot(data=expo, ax=axes[0])
sns.histplot(data=expo, kde=True, ax=axes[1])
sns.kdeplot(data=expo, ax=axes[2])

sns.displot(data=expo, kde=True)     ### `displot` is a figure-level function and does not accept the ax= parameter. You may wish to try histplot.
```

```python
expo = random.exponential(size=1000)
sns.displot(expo, hist=False)
```
