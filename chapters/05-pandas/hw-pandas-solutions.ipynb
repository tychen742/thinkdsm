### 1.1. for loop with list

```python
### create a list following the instructions below

lst1 = list(range(1, 10))       ### use the range() function to create the a list called lst1
lst2 = []                       ### create an empty list, lst2, to save the results  
for i in lst1:                  ### a for loop to loop through the elements in lst1 
    if i % 2 == 0:              ### if the element is even
        lst2.append(i**2)       ### append the square of the element to lst2
                                ###
print(lst2, end=" ")            ### print the results in lst2; use the end= parameter when printing   
```

### 1.2. for loop with list comprehension

```python
### create a list (lst3) using list comprehension following the instructions below

lst3 = [i**2 for i in lst1 if i % 2 == 0]   ### create a list comprehension to generate the same results as lst2
print(lst3, end=" ")                        ### print the results in lst3; use the end= parameter when printing
```

## 2. NumPy Array Operations

### 2.1 NumPy Vector

```python
### create a numpy array, call it arr1, from the list lst1
arr1 = np.array(lst1)           ### create a numpy array from lst1
print(arr1)                     ### print the array
print(type(arr1))               ### print the data type of the array
print(arr1.dtype)               ### print the dtype of the array elements
print(arr1.ndim)                ### print the number of dimensions of the array
print(arr1.shape)               ### print the shape of the array
print(arr1[5])                  ### print the element at index 5 (the 6th element)
print(arr1 + arr1)              ### 
print(arr1 * 5)
```

### 2.2. NumPy Matrix

```python
### create a numpy vector called v using the arange function
### v should contain integers from 0 to 19 

v = np.arange(0, 20)            ### create vector v
arr2 = v.reshape(4, 5)          ### create arr2 to a 4 x 5 matrix use reshape function
print(arr2)                     ### print the 2D array
print(type(arr2))               ### print the data type of the matrix
print(arr2.dtype)               ### print the data type of the array elements
print(arr2.ndim)                ### print the number of dimensions of arr2
print(arr2.shape)               ### print the shape of arr2
print(arr2[0:2, 3:])            ### print a subset of the matrix
```
