# Pandas Series

The first main data type you will learn about for pandas is the **Series** data type. Let's import Pandas and explore the Series object.

A Series is very similar to a NumPy array (in fact, it is built on top of the NumPy array object). What differentiates the NumPy array from a Series is that a Series can have **axis labels**, meaning it can be indexed by a label, instead of just a number location. It also doesn't need to hold numeric data; it can have any arbitrary Python Object.

A **Pandas Series** can be created by loading data from existing storage, such as a .  
It can also be built directly from **Python objects** like **lists**, **dictionaries**, or even **single scalar values**.

Let's explore this concept through some examples:

```python
import numpy as np
import pandas as pd
```

## Creating a Series

You can convert a list,numpy array, or dictionary to a Series (note that Series allows different data types). The `Series` can intake various arguments such as **data**, **index**, and **dtype**. You can specify the parameter names so that the order of the arguments supplied will not matter. 

```python
ser = pd.Series([1, 3, 5, np.nan, 6, 8])
ser
```

From the above evaluation outcome, you notice that each of the elements now has a corresponding index. This is just like lists but explicit. 

Let's also check the data type of the Series object:

```python
type(ser)
```

Now let's prepare the following data:

```python
labels = ['a','b','c']          ### a string list
lst = [10,20,30]                ### a numerical list
arr = np.array(lst)             ### a numpy array
dict = {'apple':10,'banana':20,'cherry':30}   ### a dictionary: key-value pairs
```

### From Python Lists

When supplying only the data to create a Pandas Series, indexes will be automatically created.  

```python
### feed the data

pd.Series(data=lst)
```

Usually you would want to supply a more descriptive **labels** by sending in a label list. 

```python
### add labels

pd.Series(data=lst,index=labels)
```

```python
### omit the parameter names

pd.Series(lst,labels)
```

When we need to create a Python list object in a workflow, list comprehension allows us to do so without interruption. Note that in the code below we place the list comprehension in the Series function. 

```python
### create a list using list comprehension
### this is a demonstration of list comprehension and we save the list to a variable
nums_lst = [ num for num in range(5) ]    ### [0, 1, 2, 3, 4]

### directly put the list comprehension in the Series function
pd.Series([ num for num in range(5) ])


### indexes are added automatically
```

We can turn a NumPy array into a Pandas Series.

### From Python dictionary

Since a Python dictionary maps **keys** to **values**, passing it to pd.Series will use the dictionary keys as the Series index and the corresponding values as the Series data — a convenient way to create a labeled Series from key–value pairs.

## Data types in a Series


A Pandas Series can hold data of various types, including integers, floats, strings, and even Python objects. The data type of the Series is inferred from the input data, but it can also be explicitly specified using the `dtype` parameter when creating the Series. This flexibility allows for efficient storage and manipulation of heterogeneous data within a single Series object.

Examine the examples below to learn about common Pandas Series dtypes and simple conversions/inspections:


```python
s_int = pd.Series([1, 2, 3])
s_float = pd.Series([1.0, 2.5, np.nan])
s_str = pd.Series(['a', 'b', 'c'])
s_mixed = pd.Series([1, 'a', 3.0])
s_bool = pd.Series([True, False, True])
s_dt = pd.Series(pd.date_range('2020-01-01', periods=3))
s_cat = pd.Series(pd.Categorical(['low', 'medium', 'high', 'low'],
                                 categories=['low', 'medium', 'high'],
                                 ordered=True))
s_obj = pd.Series([{'x': 1}, [1, 2], (3, 4)])

series_list = [
    ("integer", s_int),
    ("float (with NaN)", s_float),
    ("string", s_str),
    ("mixed (object)", s_mixed),
    ("boolean", s_bool),
    ("datetime", s_dt),
    ("categorical (ordered)", s_cat),
    ("object (python objects)", s_obj),
]

for name, s in series_list:
    print(f"{name:25} dtype -> {s.dtype}")

# Small demos of conversions / operations
print("\nConversions / operations:")
print("s_float -> fillna(0) -> astype(int):", s_float.fillna(0).astype(int).tolist())
print("s_str -> astype('category') dtype:", s_str.astype('category').dtype)
print("s_mixed dtype (inference):", s_mixed.dtype)
print("\ns_cat value counts (preserves categorical ordering):")
print(s_cat.value_counts(sort=False))
```

## Series Indexing

The key to working effectively with a Pandas **Series** is understanding its **index**. Pandas uses these index labels—whether they are names or numbers—to allow for fast and efficient data lookups, much like a hash table or **dictionary**.

### Accessing with indexes
Let’s explore some examples of how to access data from a Series. We will begin by creating two Series objects, ser1 and ser2:

```python
### creating a Series with custom indexes

ser1 = pd.Series( [ 10, 20, 30 ], index = [ 'Apple', 'Banana', 'Cherry' ])
ser1
```

```python
ser2 = pd.Series( [ 10, 20, 40 ], index = ['Apple', 'Banana', 'Cranberry'] )
ser2
```

```python
print(ser1['Apple'])
```

Observe the operation below. See that operations are done based off of index. Whe we add two series, Pandas aligns the two Series by their index labels before performing the operation:
- The result's index is the union of the two input indexes.
- For labels present in both Series, values are added element-wise.
- If a label is missing in one Series, the result for that label is **NaN** (missing value propagates).
- Presence of NaN may change dtype (e.g., integers -> floats) because NaN is a float value.



```python
ser1 + ser2
```

### More on indexing

Pandas Series indexing provides multiple, flexible ways to access and manipulate Pandas Series data. These tools let you select, update, and combine Series values efficiently while preserving or controlling index semantics.: 
- `.loc`: label-based access with .loc (inclusive slices), 
- `.iloc`: position-based access with .iloc (Python-style slices), 
- `.at` and `.iat`: fast scalar access/assignment with .at and .iat, 
- boolean masks and `.isin`: filtering, and reindexing/alignment methods to handle differing indexes

```python
# print("ser1:\n", ser1, "\n")
print(ser1)
```

```python

### 1) Label-based access (.loc) - inclusive slicing for labels
print(ser1.loc['Banana'], "\n")
print(ser1.loc[['Apple', 'Cherry']], "\n")
print(ser1['Apple':'Cherry'])
```

```python
### 2) Position-based access (.iloc) - integer positions, Python-style slices (end exclusive)
print(ser1.iloc[1] , "\n")
print(ser1.iloc[0:2])
```

```python
# 3) Fast scalar access / assignment (.at and .iat)
print(ser1.at['Banana'])
ser1.at['Banana'] = 25               ### update set by label
print(ser1.at['Banana'])
print(ser1.iat[0], "\n")
```

```python
### 4) Boolean indexing and isin
print(ser1[ser1 > 15], "\n")
print(ser1[ser1.index.isin(['Apple', 'Cranberry'])], "\n")
```

```python
### 5) Safe get with default (avoids KeyError)
### n pandas, the .get() method for a Series is a safe way to access values by 
### index label without raising a KeyError if the label doesn't exist. 
### It works similarly to Python's dictionary .get() method.

print(ser1.get('Durian', 'Not found'), "\n")
```

```python
### 6) Reindexing and filling missing values
new_idx = ['Cherry', 'Apple', 'Durian']
print("Reindex to", new_idx, "with fill_value=0:\n", ser1.reindex(new_idx, fill_value=0), "\n")
```

```python

### 7) Aligning / combining two Series (useful when indexes differ)
### ser1 + ser2 (alignment on index)
print(ser1 + ser2, "\n")
### fill missing in ser1 from ser2 using combine_first
print(ser1.combine_first(ser2))
```

```python

### 8) Conditional selection with where (keeps values that satisfy condition)
### "Keep values > 15, else 0 
print(ser1.where(ser1 > 15, other=0), "\n")
```

```python
### 9) Inspect index and values directly
print(ser1.index)
print(ser1.values)
```
