# Pandas Series

The first main data type you will learn about for pandas is the **Series** data type. Let's import Pandas and explore the Series object.

A Series is very similar to a NumPy array (in fact, it is built on top of the NumPy array object). What differentiates the NumPy array from a Series is that a Series can have **axis labels**, meaning it can be indexed by a label, instead of just a number location. It also doesn't need to hold numeric data; it can have any arbitrary Python Object.

A **Pandas Series** can be created by loading data from existing storage, such as a .  
It can also be built directly from **Python objects** like **lists**, **dictionaries**, or even **single scalar values**.

Let's explore this concept through some examples:

```python
import numpy as np
import pandas as pd
```

## Creating a Series

You can convert a list,numpy array, or dictionary to a Series (note that Series allows different data types):

```python
ser = pd.Series([1, 3, 5, np.nan, 6, 8])
ser
```

From the above evaluation outcome, you notice that each of the elements now has a corresponding index. This is just like lists but explicit. 

Let's also check the data type of the Series object:

```python
type(ser)
```

Now let's prepare the following data:

```python
labels = ['a','b','c']          ### a string list
lst = [10,20,30]                ### a numerical list
arr = np.array(lst)             ### a numpy array
dict = {'a':10,'b':20,'c':30}   ### a dictionary
```

### List & labels

```python
### feed the data

pd.Series(data=lst)
```

```python
### add labels

pd.Series(data=lst,index=labels)
```

```python
### omit the parameter names

pd.Series(lst,labels)
```

### NumPy Array

```python
pd.Series(arr)
```

#### list comprehension to ndarray

```python
### create a list using list comprehension
nums = [ num for num in range(5) ]    ### [0, 1, 2, 3, 4]
arr_nums = np.array(nums)             ### array([0, 1, 2, 3, 4])

ser_nums = pd.Series(arr_nums)
ser_nums

### indexes are added automatically
```

```python
### back to arr: 10, 20, 30 

pd.Series(arr,labels)
```

### Dictionary

```python
### key-value pairs

pd.Series(dict)
```

## Data types in a Series

A pandas Series can hold a variety of object types:

```python
### passing the labels as data for fun

pd.Series(data=labels)
```

```python
### even list/tuple... or functions (although unlikely that you will use this)

print(pd.Series([tuple, list, np.array]), "\n")
print(pd.Series([sum,print,len]))
```

## Series Indexing

The key to working effectively with a Pandas **Series** is understanding its **index**. Pandas uses these index labels—whether they are names or numbers—to allow for fast and efficient data lookups, much like a hash table or dictionary.

Let’s explore some examples of how to access data from a Series. We will begin by creating two Series objects, ser1 and ser2:

```python
### 

ser1 = pd.Series([1, 2, 3], index = ['Apple', 'Banana', 'Cherry'])                                   
```

```python
ser1
```

```python
ser2 = pd.Series([1, 2, 4], index = ['Apple', 'Banana', 'Cranberry'])
```

```python
ser2
```

```python
print(ser1['Apple'])
```

Operations are then also done based off of index:

```python
ser1 + ser2
```
