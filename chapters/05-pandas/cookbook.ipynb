```python
df = pd.DataFrame(
    {
        "AAA": range(4, 8), 
        "BBB": np.arange(10, 50, 10), 
        "CCC": np.random.randint(10, 15, 4)
    }
)
df
```

### if-then...

- An if-then on one column
- assignment changes the df

```python
df.loc[df.AAA >= 5, "BBB"] = -1     ### 
df
```

if-then on 2 columns

```python
df.loc[df.AAA >=5, ["BBB", "CCC"]] = 555
df
```

Add another line with different logic, to do the -else

```python
df.loc[df.AAA < 5, ["BBB", "CCC"]] = 2000
df
```

Or use pandas where after you’ve set up a mask



```python
df_mask = pd.DataFrame(
{
    "AAA": [True] * 4,
    "BBB": [False] * 4,
    "CCC": [True, False] * 2
}
)

df_mask
```

```python
### use the mask

df[ df_mask ]
```

```python
### use where()

df.where(df_mask, -1000)
```
