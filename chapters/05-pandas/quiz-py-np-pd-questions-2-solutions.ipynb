### 1. for loop with function

```python
### create a list following the instructions below
### using list comprehension OR a for loop
### a for loop will take more lines of code, so I leave more space for that
### you can choose either method; name it square_evens

### list comprehension version
def square_evens(nums):             ### define function; takes a list as input
    return [num for num in nums if num%2 == 0]      ### return a list of squares of even numbers

### for loop version
def square_evens(nums):             ### define function; takes a list as input
    results = []                    ### create an empty list to store results
    for num in nums:                ### iterate through each number in the input list
        if num % 2 == 0:            ### check if the number is even
            results.append(num**2)  ### if even, append its square to the results list
    return results                  ### return the list of squared even numbers

nums = list( range(1, 10) )         ### create a list of numbers from 1 to 9       
results = square_evens(nums)        ### call the function with the list

for num in results:                 ### iterate through the results
    print(num, end=" ")             ### print each squared even number followed by a space   
```

### 2.2. NumPy Matrix

```python
### create a numpy vector called v using the arange function
### v should contain integers from 0 to 19 

v = np.arange(20)           ### create vector v
arr2 = v.reshape(4, 5)      ### create arr2 to a 4 x 5 matrix use reshape function
print(arr2)                 ### print the 2D array
print(arr2.dtype)           ### print the data type of the matrix
print(arr2.dtype.type)      ### print the data type of the array elements
print(arr2.ndim)            ### print the number of dimensions of arr2
print(arr2.shape)           ### print the shape of arr2
print(arr2[1, 2])           ### print a subset of the matrix
```

### 3.3. Create Pandas DataFrame

```python
### create a Pandas DataFrame, call it df, using arr2
### use the following column names: A, B, C, D, E

df = pd.DataFrame(arr2, columns=['A', 'B', 'C', 'D', 'E'])  ### create the DataFrame
print(df)                                                   ### print the DataFrame    
```

### 3.5. Missing Data


```python
### create a DataFrame, call it df, with missing data
### use the following data

df = pd.DataFrame({                     ### create the DataFrame
    'A' : [ 1, 2, np.nan ],             ### np.nan means "Not a Number", i.e., missing data
    'B' : [ 5, np.nan, np.nan],         
    'C' : [ 1, 2, 3]
    })
```

```python
print(df)                                     ### print the DataFrame
```

```python
### detect the missing data in DataFrame df using the isna() method

print(df.isna(), "\n")          ### detect the missing data in DataFrame df using the isna() method
print(df.isna().sum(), "\n")    ### print the number of missing data in each column
print(df.fillna(0), "\n")       ### fill the missing data with 0 and print the DataFrame
```
