## Read Data Files 

### About path

When importing datasets, some path information are good to know: 
- Data paths: 
  - For macOS (and Linux), a full path starts with "`/`" (the "root") or you can use "`~`" to start with the user's home directory (e.g., "/Users/<user>`/`workspace/data/" == "`~`/workspace/data/"). 
  - In Windows, a full path includes a drive letter and backslashes for path separator (e.g. "`C:`\Users\<user>\workspace\dsm\data\") though forward slashes usually work as well.
- `CWD` (current working directory): You, from time to time, need to know your current working directory (or, where this Jupyter Notebook's location is), in order to access other files/directories from this CWD. To get your CWD, do `CWD = os.getcwd()`. Then, when you need to access `workspace/dsm/data`, you need only to start with CWD. 
- `os.path.expanduser`: To write portable code across OSs, we prefer `os.path.expanduser`, which can solve the directory passed (in our case, usually `~`) 
- `os.path.join`: then we use `os.path.join` instead of hardcoding separators or literal paths to point to the path_folder and data file.
- Relative paths; Sometimes it's just easier to use relative path to access the data folders (`../../data/`). 

````python
import os               ### os is a standard library module that provides functions for interacting with the OS.
CWD = os.getcwd()
CWD                     ### print current working directory
                        ### this is where this Jupyter Notebook is located
### note this is the same as %pwd in a Jupyter Notebook
### this is the location of the Jupyter Notebook
### we will use this to access other files/folders
### relative to this location
````

### Path to Dataset Files

Use `os.path.join( )` to join a path and a dataset file.

````python
### we want to access the data directory so that we can read data files 
### without specifying the full path
### os.path.expanduser can be used to

### you can expand ~ to the full user home path and keep the path as a string
### in the terminal/shell, it's $HOME but in Python it's os.path.expanduser('~')
home_dir = os.path.expanduser('~')
data_dir = home_dir + '/workspace/dsm/data/'
if os.path.isdir(data_dir):		### check if the directory exists
	data_file = os.path.join(data_dir, 'data.csv')
 
data_file
### this is the full path to the data file data.csv 
### we can use this path to read the data file
### without worrying about the OS path separator
````

Now let us read the dataset file using `pd.read_csv( )`. You may open the data file and save it a variable so you can manipulate the data file.

````python
### now we can read the data file using pandas
### this is one way to do it
with open(data_file) as f:      ### open the file for reading and assign the file object variable f
    df = pd.read_csv(f)         ### read the csv file into a DataFrame
df.head()                       ### display the first 5 rows of the DataFrame
````

Otherwise, just read the file:

````python
### another way to do it is to use relative path
### ../../data/data.csv
data = ("../../data/data.csv")
df = pd.read_csv(data)
df.head()
````
