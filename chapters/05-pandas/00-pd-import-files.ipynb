## Accessing Data Files

To access datasets (usually files), you can either: 
- set a path variable to the data folder, which will be easier in the long run; or
- write path (full or relative) and file names as arguments

You are learning these basic operations now and will probably just use **relative path** in the future.

### About path

When importing datasets, some path information are good to know: 
- Data paths: 
  - For macOS (and Linux), a full path starts with "`/`" (the "root") or you can use "`~`" to start with the user's home directory (e.g., "/Users/<user>`/`workspace/data/" == "`~`/workspace/data/"). 
  - In Windows, a full path includes a drive letter and backslashes for path separator (e.g. "`C:`\Users\<user>\workspace\dsm\data\") though forward slashes usually work as well.
- Current working directory (CWD)): You, from time to time, need to know your current working directory (or, where this Jupyter Notebook's location is), in order to access other files/directories from this CWD. To get your CWD, run `os.getcwd()`. 
- Home directory: To write portable code across OSs, we prefer `os.path.expanduser`, which can resolve the directory passed (in our case, usually `~`) for us to use. 
- `os.path.join`: This is used to join the path variable and the filename instead of hardcoding separators and literal paths, i.e., use `os.path.join( )` to join a path and a dataset file.
- Relative paths; Sometimes it's just easier to use relative path to access the data folders (e.g., `../../data/`). 

As an example, the `os.getcwd()` function will give us the full path of the notebook you are working on.

````python
import os               ### os is a standard library module that provides functions for interacting with the OS.
CWD = os.getcwd()
CWD                     ### print current working directory
                        ### this is where this Jupyter Notebook is located
### note this is the same as %pwd in a Jupyter Notebook
### this is the location of the Jupyter Notebook
### we will use this to access other files/folders
### relative to this location
### the result will vary different environments
````

### Reading Dataset Files

Now let us read the dataset file using `pd.read_csv( )`. You may read the data file directly or and open it as a file and save it a variable so you can manipulate the data file.

````python
### now we can read the data file using pandas
### this is one way to do it
home = os.path.expanduser('~')
data_path = os.path.join(home, 'workspace', 'dsm', 'data')
data = os.path.join(data_path, 'data.csv')
with open(data) as f:      ### open the file for reading and assign the file object variable f
    df = pd.read_csv(f)         ### read the csv file into a DataFrame
df.head()                       ### display the first 5 rows of the DataFrame
````
