## Reading Dataset Files

Once you have access to the (.csv) file, you can:

1. load the CSV file directly             
    ```python 
    df = pd.read_csv(data_file)          ### we usually use this method to read a file
    ```
    
2. open file: with **open( )**:
   ```python
   file = open(data_file, "r")           ### r == read; after done with the file, you need to close it
   df = pd.read_csv(file)
   ```
   
3. open file: using **with** statment:        
   ```python                               
   with open(data_file, "r") as file:    ### the with statement will close the file automatically 
       df = pd.read_csv(file)
    ```

````python
### I encourage you to learn how to do this

with open(data) as f:                        ### you have the file variable f to use if you need to
    df = pd.read_csv(f)                      ### we usually just work with df
````

## Creating DataFrames

Creating a DataFrame in Pandas can be done in several ways:

  1. From a Python **Dictionary** (Most Common)
  2. From a **List** of Dictionaries
  3. From a List of Lists (with **Column** Names)
  4. From **Numpy Arrays**
  5. Create an Empty DataFrame

Check DataFrame Info

### Python dictionary to df

````python
data = {
    "Name": ["Johnson", "Kylee", "Hector", "Cameryn"],        ### key : [ values ]
    "Age": [25, 30, 35, 24],                           ### key : [ values ]
    "City": ["Rolla", "St. Louis", "Chicago", "St. James"]      ### key : [ values ]
}

### Create DataFrame: key-value pair ==> Series (columns)

df = pd.DataFrame(data)
df
````

### List of Dictionaries to df

````python
### 4 dictionaries in the list

data = [
    {"Name": "Johnson", "Age": 25, "City": "Rolla"},
    {"Name": "Kylee", "Age": 30, "City": "St. Louis"},
    {"Name": "Hector", "Age": 35, "City": "Chicago"},
    {"Name": "Cameryn", "Age": 24, "City": "St. James"}
]

df = pd.DataFrame(data)
df
````

### List of lists to df

````python
data = [
    ["Johnson", 25, "Rolla"],
    ["Kylee", 30, "St. Louis"],
    ["Hector", 35, "Chicago"],
    ["Cameryn", 24, "St. James"]
]

columns = ["Name", "Age", "City"]

pd.DataFrame(data, columns = columns)
````

### NumPy Arrays to df

If you are working with NumPy, you can pass an array directly.

````python
data = np.array([
    ["Johnson", 25, "Rolla"],
    ["Kylee", 30, "St. Louis"],
    ["Hector", 35, "Chicago"],
    ["Cameryn", 24, "St. James"]
])

# pd.DataFrame(data)
pd.DataFrame(data, columns = ["Name", "Age", "City"])
````

## DataFrame from SQL Databases
Now, let us try to load the chinook SQLite database that we have used in the SQL database activity. You will be able to appreciate how convenient Pandas makes loading data to be:

1. load sqlite3 library (import; the sqlite module is part of the Python standard library)
2. connect to the database by calling the sqlite3.connect() function

## Multi-Index and Index Hierarchy

Let us go over how to work with Multi-Index, first we'll create a quick example of what a Multi-Indexed DataFrame would look like:

````python
# Index Levels
outside = ['G1','G1','G1','G2','G2','G2']
inside = [1,2,3,1,2,3]
hier_index = list(zip(outside,inside))
hier_index = pd.MultiIndex.from_tuples(hier_index)
````

````python
hier_index
````

````python
df = pd.DataFrame(np.random.randn(6,2),index=hier_index,columns=['A','B'])
df
````

Now let's show how to index this! For index hierarchy we use df.loc[], if this was on the columns axis, you would just use normal bracket notation df[]. Calling one level of the index returns the sub-dataframe:

````python
df.loc['G1']
````

````python
df.loc['G1'].loc[1]
````

````python
df.index.names
````

````python
df.index.names = ['Group','Num']
````

````python
df
````

````python
df.xs('G1')
````

````python
df.xs(['G1',1])
````

````python
df.xs(1,level='Num')
````
