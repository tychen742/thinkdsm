## Unpacking

Unpacking means assigning elements of a collection (like a list, tuple, or dictionary) to multiple variables in a single statement. In other words, Python lets you “open up” a data structure and assign its parts to separate names — all in one line.

Common upacking use cases are:

| Type                          | Syntax                 | Example                        | Common Use Case                                |
| ----------------------------- | ---------------------- | ------------------------------ | ---------------------------------------------- |
| **Tuple/List Unpacking**      | `a, b = (1, 2)`        | Assigns two values at once     | Simplify variable assignment                   |
| **Extended Unpacking**        | `a, *b = [1, 2, 3, 4]` | a = 1, b = [2, 3, 4]           | Capture “the rest” of data                     |
| **Iterable Unpacking**            | `for x, y in pairs:`   | `for name, grade in students:` | Iterate over tuples or key–value pairs |
| | `for i, val in enumerate(seq):` |for i, name in enumerate(students): | Get index + value together |
| **Dictionary Unpacking**      | `x, y = d`            | `d = {'a': 1, 'b': 2}         | keys-value pair variable assignment                   |
|       | `**dict`               | `greet(**info)`                | Pass dict keys as named args                   |
| **Function Return Unpacking** | `x, y = func()`        | `mean, std = stats(data)`      | Receive multiple return values                 |
| | | `q25, q75 = np.percentile(data, [25, 75])`|
| | | `fig, axes = plt.subplots(1, 2)`|

## Tuple Unpacking

```python
point = (4, 5)

# Traditional way
x = point[0]
y = point[1]

# Using unpacking
x, y = point

print(x)  # 4
print(y)  # 5
```

Unpacking is useful because it makes code cleaner and more readable when dealing with:

- Dataset columns
- Function returns (like mean, median, std)
- Iterating over rows or key-value pairs

## Iterable Unpacking

### Loop Unpacking

```python
coordinates = [(1, 2), (3, 4), (5, 6)]

for x, y in coordinates:
    print(f"x={x}, y={y}")
```

### Dictionary Unpacking

```python
dict = {'a': 1, 'b': 2}
x, y = dict              ### unpack keys: x='a', y='b'

def greet(name, age):
    print(f"Hello {name}, age {age}")

student = {"name": "Alice", "age": 21}
greet(**student)
```

### Unpacking with Pandas DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    'A': [9, 9, 9],
    'B': [1, 2, 3]
})

rows, cols = df.shape     ### unpack shape, which has two values
print(f"The DataFrame has {rows} rows and {cols} columns.")   ### note that A and B are columns
```
