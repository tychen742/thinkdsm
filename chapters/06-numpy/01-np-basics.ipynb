# NumPy Arrays Basics

NumPy arrays (n-dimensional array, or `ndarray`) NumPy is like a super-powered list that can store numbers in rows and columns. NumPy arrays essentially come in two flavors: vectors and matrices. Vectors are strictly 1-d arrays has only one axis, and matrices are 2-d arrays with two axes: (rows, columns).

## Creating NumPy Arrays

### From a Python List

You can create an array by directly casting a list to an array. Let us create a list first:

```python
nums_list = [1,2,3, 4, 5]
nums_list
```

```python
type(nums_list)
```

Now let's cast the list into a numpy array using numpy's `array()` method to create a **1-D** array.

```python
### before using numpy, import
### we usually import packages in the very beginning of a notebook

import numpy as np
```

```python
arr = np.array(nums_list)       ### casting a list to a numpy array
arr

### now you will see a numpy array, which is different from the list [1, 2, 3]
```

```python
### data type is different from list too: numpy.ndarray 

type(arr)                   
```

We can also place the Python list directly into the np.array function to create a NumPy array (note that this is a **2-D** array):

```python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
print(arr)
```

An example of Python **3-D** "array" (list of lists): 

```python
list_of_list = [
    [
        [1, 2, 3, 4], 
        [2, 2, 3, 4], 
        [3, 2, 3, 4]
    ], 
    [
        [4, 2, 3, 4], 
        [5, 2, 3, 4], 
        [6, 2, 3, 4]
    ]
]
list_of_list
```

Compare the result of evaluation of the Python list above with the NumPy array from the list:

```python
np.array(list_of_list)
```

Note the syntax of list comprehension is an alternative to for loops. Instead of initialize an empty list variable, running the for loop to append the elements, list comprehension is more concise. 

```python
np.array([i**2 for i in range(10)])
```

### Using Functions: zeros(), ones(), and full()
These function initialize arrays of a specified shape filled with zeros, ones, or a constant value, respectively. The syntax for numpy.ones() function is:

<code>numpy.ones(shape, dtype=None, order='C', *, like=None)</code>

We pass the shape argument using a list or tuple. 

```python
np.zeros((2, 3))        ### by default dtype is float
```

```python
np.ones([2,3], int)     ### dtype is int
```

```python
np.full((2, 3), 5)      ### fill the shape (2x3) with 5
```

## NumPy Indexing and Selection

Just like Python lists, square brackets [] are used for selecting the elements or groups of elements from an array:
1. accessing individual elements of a list (**indexing**).
2. selecting sub-sequences (**slicing** with parameters: start, stop exclusive, and step). Note that slicing returns a numpy array.
3. **negative** indexing: begins with -1 from the end of the sequence.

```python
import numpy as np
```

```python
### creating sample array
arr = np.arange(0,10)
```

```python
### show the array
arr
```

```python
### indexing 
arr[1]
```

```python
arr[0:3]
```

```python
arr[0:5:2]
```

```python
arr[-1]
```

```python
### start is -3, and stop is -1 exclusive
arr[-3:-1]
```

### Indexing/Slicing 2-D Aarrays (Matrices)

To access a 2-D array (matrices), use the syntax `array[row_index, column_index]`.

```python
arr_2d = np.array(([5,10,15],[20,25,30],[35,40,45]))
arr_2d
```

```python
### 0-based indexing; 1st row
arr_2d[0]
```

```python
### 0-based indexing
arr_2d[1, 2]
```

```python
### Slicing
arr_2d[:2, 1:]
```

### The reshape() Function 
The reshape() function changes the shape of a NumPy array without altering its underlying data. It returns a new array (or a view of the original array if possible) with the specified new shape.

```python
### reminder: arr
arr
```

```python
# Notice the two sets of brackets
arr = arr.reshape(2,5)
arr
```

```python
### list comprehension
arr = np.array([ num for num in range(25)])
arr
```

```python
arr.reshape(1,25).shape
```

```python
arr.reshape(25,1)
```

```python
arr.reshape(25,1).shape
```

## Indexing a 2D array (matrices)

The general format is **arr_2d[row][col]** or **arr_2d[row,col]**. I recommend usually using the comma notation for clarity.

```python
### creating a 2-d array

arr_2d = np.array(([5,10,15],[20,25,30],[35,40,45]))

#Show
arr_2d
```

```python
### indexing row
arr_2d[1]
```

```python
### Format is arr_2d[row][col] or arr_2d[row,col]

### Getting individual element value
arr_2d[1][0]
```

```python
### getting individual element value
arr_2d[1,0]
```

```python
# 2D array slicing

### shape (2,2) from top right corner
arr_2d[:2,1:]
```

```python
#Shape bottom row
arr_2d[2]
```

```python
#Shape bottom row
arr_2d[2,:]
```
