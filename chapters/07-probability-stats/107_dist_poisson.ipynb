```python
from numpy import random
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
random.seed(42)
pois = random.poisson(lam=2, size=1000)

print(pois)
sns.histplot(pois, kde=False, alpha=.5)
plt.show()
```

```python
random.seed(42)
norm = random.normal(loc=50, scale=7, size=1000)
pois = random.poisson(lam=50, size=1000)

# sns.displot(norm, kind='kde', label='normal', color='blue')
# sns.displot(pois, kind='kde', label='poisson', color='red')

df = pd.DataFrame({'norm': norm, 'pois': pois})
sns.displot(data=df, kind='kde')
# sns.distplot(pois, hist=False, label='poisson', color='red')
plt.show()
```
