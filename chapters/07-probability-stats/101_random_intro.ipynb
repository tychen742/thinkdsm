```python
from numpy import random
```

## random.randint()

```python
for i in range (0, 11):
    rand = random.randint(100)
    print(rand, end=" ")
```

```python
for i in range(1, 4):
    rand = random.randint(0, 10)
    print(rand, end="\t")
    # print(i)
```

## random.rand()

```python
### random float: rand()
for i in range(1, 4):
    rand_float = random.rand()
    print(rand_float)
print()

for i in range(1, 4):
    rand_float = random.rand(2)
    print(rand_float)
print()

for i in range(1, 4):
    rand_float = random.rand(3, 2) # one row, each row contains 4 random numbers
    print(rand_float)
```

```python
for i in range(1, 6):
    rand_float = random.rand()
    print(rand_float)
```
