## bins

As I mentioned a couple of times before, `bins` is tricky. Look at the example below.

`sns.displot` (a wrapper over histplot) treats your list as **continuous data** and chooses automatic **bin edges**. With only 6 integers, the auto binning groups multiple values into the same bin (e.g., [0,1], [2], [3], [4,5]), so the largest bin contains 2 points, hence the tallest bar is at count = 2.

```python
### plotting a distplot w/o the histogram

lst = [0, 1, 2, 3, 4, 5]
sns.displot(lst, height=3, aspect=5/3)
# plt.show()
# plt.tight_layout()
```

```python
fig = plt.figure(figsize=(5, 3))
# fig.set_size_inches(3, 2)


# g = sns.displot(lst, bins=6, height=3, aspect=5/3)
# g.fig.subplots_adjust(left=0.1, right=0.95, top=0.95, bottom=0.1)

sns.displot(lst, discrete=True, height=3, aspect=5/3)             # treats x as discrete categories
sns.displot(lst, binwidth=1, height=3, aspect=5/3)                # 1-wide bins
sns.displot(lst, bins=6, height=3, aspect=5/3)                    # exactly 6 bins
plt.tight_layout()
```
