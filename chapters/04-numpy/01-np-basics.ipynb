# NumPy Arrays Basics

NumPy arrays (n-dimensional array, or **`ndarray`**) are like a super-powered **list** that can store numbers in rows and columns. You will learn that some array operations are similar, if not the same, as Python lists.

NumPy arrays essentially come in two flavors, vectors and matrices:
- **Vectors** are strictly 1-d arrays that have only one axis, and
- **Matrices** are 2-d arrays with two axes: (rows, columns).

Here in this section, we discuss topics regarding NumPy arrays:
- creation
- indexing/slicing
- general array attributes and functions such as ndim, dtype, shape, reshape()


## NumPy Data Structure

The primary data structure in NumPy is the N-dimensional array, or **`ndarray`**. NumPy's arrays are a list of lists in Python, but are more compact than Python lists. In essence, a Python list is an array of pointers to heterogeneous Python objects, while a NumPy array is an array of **uniform** values of the same type (e.g., all integers, all floats) and array elements are stored in one continuous block of memory, similar to how arrays work in C ({numref}`c-vs-python-list`). Python lists are more flexible, but Numpy arrays are smaller in file size, and access in reading and writing items is much faster {cite}`Martelli_2009`.

## Creating NumPy Arrays

### From a Python List/Tuple

You can create an array by directly **casting** a list to an array. Let us create a list first:

````python
nums_list = [1,2,3, 4, 5]
nums_list
````

````python
type(nums_list)
````

Now let's cast the list into a numpy array using numpy's `array()` method to create a **1-D** array.

````python
### before using numpy, import
### we usually import packages in the very beginning of a notebook

import numpy as np
````

````python
arr = np.array(nums_list)       ### casting a list to a numpy array
arr

### now you will see a numpy array, which is array([ ele, ele, ele ]) from the list [1, 2, 3, 4, 5]
````

````python
### data type is different from list too: numpy.ndarray 

type(arr)                   
````

We can also place the Python list directly into the np.array function to create a NumPy array (note that this is a **2-D** array):

````python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
print(arr)
````

An example of Python **3-D** "array" (list of lists):

### Using arange() Function

**`numpy.arange()`** is a NumPy function used to create arrays with evenly spaced values with a specified interval. It is similar to Python's built-in `range()` function, but returns a NumPy `ndarray` instead of a `range` object, which is more efficient for numerical operations.


````python
np.arange(5)
````

Using the range parameters: start, stop exclusive, stop:

````python
np.arange(2, 11, 2)
````

## Using the numpy.random module 

(These functions are considered "[legacy random generation](https://numpy.org/doc/stable/reference/random/legacy.html#legacy)" and are for simple random data. 

The numpy.random module in NumPy provides tools for generating pseudo-random numbers and sampling from various probability distributions. Some functions in the numpy.random module are commonly used:
- **np.random.rand()**: returns random floating-point numbers (uniform distribution over the interval [0, 1))
- np.random.randn(): returns random floating-point numbers ("standard normal" distribution with mean of 0 and standard deviation of 1)
- **np.random.randint()**: returns random integers from a specified range
- **np.rand
- (note that np.random.random is the same as random.rand(), except random.random() takes the shape argument as a single tuple; e.g., see [stackoverflow](https://stackoverflow.com/questions/47231852/np-random-rand-vs-np-random-random))

````python
### rand: one number

np.random.rand()
````

````python
### rand: 1D array

from numpy import random     ### different import syntax

random.rand(3)
````

````python
### rand: 2D

random.rand(2, 3)
````

````python
### randint: 1 number

print(random.randint(10))   ### range between 0 and 10 exclusive
print(random.randint(5, 10))   ### range between 5 and 10 exclusive
````

````python
### randin: 1D array (with size parameter)

random.randint( 0, 10 )   ### 0 to 10 exclusie
````

````python
### the choice function

random.choice([3, 5, 7, 9], size = 2)
````

## NumPy Indexing and Selection

Just like Python lists, square brackets [] are used for selecting the elements or groups of elements from an array:
1. accessing individual elements of a list (**indexing**).
2. selecting sub-sequences (**slicing** with parameters: start, stop exclusive, and step). Note that slicing returns a numpy array.
3. **negative** indexing: begins with -1 from the end of the sequence.

````python
import numpy as np
````

````python
### creating sample array
arr = np.arange(0,10)
````

````python
### show the array
arr
````

````python
### indexing 
arr[1]
````

````python
arr[0:3]
````

````python
arr[0:5:2]
````

````python
arr[-1]
````

````python
### start is -3, and stop is -1 exclusive
arr[-3:-1]
````

### Indexing/Slicing 2-D Aarrays (Matrices)

To access a 2-D array (matrices), use the syntax `array[row_index, column_index]`. The general format is **arr[row][col]** or **arr[row, col]**. 

````python
### creating a 2-d array

arr_2d = np.array(([5,10,15],[20,25,30],[35,40,45]))
arr_2d
````

````python
### 0-based indexing; 1st row
arr_2d[0]
````

````python
### 0-based indexing
arr_2d[1, 2]
````

````python
### 2D array slicing
### shape (2,2) from top right corner

arr_2d[:2, 1:]
````

#### One More Example: 2-D Array (Matrices)

````python

arr_2d = np.array(([5,10,15],[20,25,30],[35,40,45]))

#Show
arr_2d
````

````python
### indexing row
arr_2d[1]
````

````python
### Format is arr_2d[row][col] or arr_2d[row,col]

### Getting individual element value
arr_2d[1][0]
````

````python
### getting individual element value
arr_2d[1,0]
````

````python
# 2D array slicing

### shape (2,2) from top right corner
arr_2d[:2,1:]
````

````python
### shape bottom row
arr_2d[2]
````

````python
### also shape bottom row with colon
arr_2d[2,:]
````

````python
arr.ndim
````

The shape attribute **returns a tuple** of integers at every index tells about the number of elements the corresponding dimension has.

````python
arr_2d = np.array([ [1, 2, 3], [4, 5, 6] ])
arr_2d.shape
````

````python
### create a 1-D array

arr_vec = np.arange(1, 10, 2)
arr_vec
````

### The reshape() Function 
The reshape() function changes the shape of a NumPy array without altering its underlying data. It **returns a new array** (or a view of the original array if possible) with the specified new shape.

````python
### arr
arr = np.array(
    [num for num in range(16)]
)

arr
````

````python
### reshape
arr = arr.reshape(4, 4)
arr
````

````python
arr.reshape(2, 8)
````

````python
### this will not work

# arr.reshape(5,5)
````

````python
arr_25 = np.arange(25)
arr_25.reshape(5, 5)
````

````python
arr_25.reshape(5, 5).shape
````

````python
### reshape does not change the object

arr_25
````
