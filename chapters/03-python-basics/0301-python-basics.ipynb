## Variables

**Variables** are names that reference **objects** in memory. A couple of things to know about Python variables include:

- You create variables with the assignment operator (the equal sign, e.g., x = 2) and
- Can **reassign** them to different objects.
- Use descriptive, **lowercase** names with **underscores** for readability, and
- Avoid starting names with digits or symbols.

For naming conventions, see [PEP8](https://peps.python.org/pep-0008/) or [Google Python Style Guide](https://google.github.io/styleguide/pyguide.html).

### Creating Variables

Variables are created by assigning values through the use of the equal sign. 

`````python
num1 = 100
num2 = 200
num3 = num1 + num2          ### reassign with expression
num3 = num3 + num3          ### reassign

university = "S&T"          ### string object variable

user_age = 30               ### lower-case with underscore
total_volume = 15.5
is_active = True

MAX_CONNECTIONS = 100000    ### UPPER case for constant
PI = 3.14159
`````

`````python
### illegal names

# my-var = 10
# my var = 10
# 2myvar = 10

# 12var = 12
# $var = 5
`````

### Python Objects
- Everything in Python is an **object**.
- A variable doesn't "contain" a value; it **points** to an object in memory.
- Objects come from **classes**, which act like blueprints.
- Each object has:
  - **Identity** (where it lives in memory)
  - **Type** (what kind of thing it is)
  - **Value** (the data it holds)

To check the identy and type of an object, we can use the `id()` and `type()` method:

`````python
num = 100                                                            ### assign a value to a variable name/label
print("The unique identifier of this num object is: ", id(num))      ### identity of object
print("The type/class of this num object is: ", type(num))           ### data type of object
`````

Further, we know that Python objects have attributes and methods. Python objects encapsulate variables (attributes) and methods (functions) that operate on those variables: 
  - **Attributes**: Variables associated with an object or a class.
  - **Methods**: Functions defined within a class that describe the object's behavior. 

To access the attributes and methods of a Python object, we use the **dot operator** after its name. For example: 

`````python
university = 'MS&T'

print("The num object is an integer: ", num.is_integer())
print("The word object is a decimal: ", university.isdecimal())
print("Turn word to lower case: ", university.lower())
`````

Note that we use the `dir()` function to show all the attributes and methods for the object. 

`````python
print(dir(word))
`````

## Operators

**Operators** enable you to perform common operations on variables. There are seven main operator types:

| Operator type | Purpose | Examples |
| --- | --- | --- |
| Assignment | Perform assignments and augmented assignments | `=`, `+=` |
| Arithmetic | Perform calculations (note `/` yields a floating-point number; avoid division by 0) | `+`, `-`, `*`, `/`, `%`, `//`, `**` |
| Comparison | Evaluate relationships | `==`, `!=`, `<`, `>`, `<=`, `>=` |
| Logical | Combine boolean expressions | `and`, `or`, `not` (e.g., `x == x and y > z`) |
| Identity | Compare object identity; equality compares values | `is`, `is not` (e.g., `a is b`) |
| Membership | Test membership in a collection | `in`, `not in` (e.g., `x in my_list`) |
| Bitwise | Bit-level operations | `&`, `|`, `^`, `<<`, `>>` (e.g., `x & y`) |

### Operator Precedence

Python follows standard mathematical order of operations (PEMDAS: Parentheses, Exponents, Multiplication and Division from left to right, then Addition and Subtraction from left to right). For example:

`````python
print(2 + 3 * 5 + 5)
print(2 + 3 * (5 + 5))
`````

### Numbers

Python has three basic number types: integers (e.g., 1), floating-point numbers (e.g., 1.0), and complex numbers. The standard order of operations for basic arithmetic is followed. Note that dividing two integers results in a floating-point number, and dividing by zero will generate an error.

`````python
1 + 1         
1 * 3         
1 / 2         
2 / 2             ### output 1.0, not 1
2 ** 4        
2 + 3 * 5 + 5     ### 22
2 + 3 * (5 + 5)   ### 32
`````


The modulus operation (also known as the mod function) is represented by the percent sign. It returns what remains after the division:



`````python
4 % 2
5 % 2
8 % 2
`````


### Strings

Strings can be created using single or double quotes. You can also wrap double quotes around single quotes if you need to include a quote inside the string.



`````python
'hello'
"hello"
"I can't go"
`````


#### Indexing and Slicing Strings

Strings are **sequences** of characters. You can access specific elements using **square bracket notation**. Python indexing starts at zero.



`````python
s = 'hello'
s[0]
s[4]
`````


**Slice notation** allows you to grab parts of a string. Use a colon to specify the start and end indices. The end index is not included.



`````python
s = 'abcdefghijk'
s[0:]
s[:3]
s[3:6]
`````


### Lists

Lists are sequences of elements (similar to the array in many other languages) in **square brackets**, separated by _commas_. Lists can contain any data type, including other lists.



`````python
my_list = ['a', 'b', 'c']
my_list.append('d')
my_list[0]
my_list[1:3]
my_list[0] = 'NEW'
`````


Lists can be nested inside each other. You can access elements in nested lists by chaining square brackets.



`````python
lst = [1, 2, [3, 4]]
lst[2][1]             ### 4
`````


For deeper nesting, continue stacking brackets to access the desired element.



`````python
nest = [1, 2, 3, [4, 5, ['target']]]
nest[3][2][0]                         ### 'target'
print(nest[3][2][0])                  ### target
`````

## Input and Output

In Python, basic input and output operations are handled primarily by the input() and print() functions.

### input()

The input() function is used to obtain user input from the command line. It displays a prompt for the user to enter text and press Enter. The function then returns the entered text as a string.



`````python
name = input("Enter your name: ")
print(f"Hello, {name}!")
`````


```
>>> num = input("Enter an integer between 1 and 10: ")
Enter an integer between 1 and 10: 7
>>> num
'7'
>>> type(num)
<class 'str'>
>>>
```

### print()

The print function is a Python built-in function that output text. In Jupyter Notebook, typing the variable name displays the value, but to officially display output in Python, we use the `print` statement.

You can pass a string **literal** directly to print() as an argument:



`````python
>>> print("Please wait while the professor is thinking...")
Please wait while the professor is thinking...
`````


Mostly, you pass **variables** to print() instead of a string literal:



`````python
greeting = 'hello world'
print(greeting)
`````


You can also pass **expressions** to print():



`````python
>>> x = 10
>>> y = 2
>>> print( x**y )
100
>>>
`````


print() also accepts **multiple arguments** separated by **comma**:


`````python
>>> name = "Dr. Chen"
>>> age = 35
>>> print(name, age)
Dr. Chen 35
>>> 
`````


You can also mix **expressions** with literals when printing using commas (a space between each item will be added):


`````python
name = "Bob"
age = 25
print("Hello,", name, "! You are", age, "years old.")
`````


If you use **concatenation** (`+`) to join the arguments, you have to cast the arguments to string (note that concatenation does not add commas between arguments):



`````python
>>> print(name, age)
Dr. Chen 35
>>> print(name + age)
Traceback (most recent call last):
  File "<python-input-4>", line 1, in <module>
    print(name + age)
          ~~~~~^~~~~
TypeError: can only concatenate str (not "int") to str
>>> print(name + str(age))
Dr. Chen35
>>> 
`````


**F-strings**, also called formatted string literals, are string literals that have an `f` before the opening quotation mark. They can include Python expressions enclosed in **curly braces**. Python will replace those expressions with their resulting values. F-strings are more flexible and readable and therefore is in general preferred for string formatting: 

```python
>>> name = "Dr. Chen"
>>> age = 35
>>> print(f"Hi, {name}, you are {age} years old!")
Hi, Dr. Chen, you are 35 years old!
>>> 
```

The `end` and `sep` parameters in the print function are used to control the output format. `end` (default to `\n`, the newline character) controls what character(s) are printed at the end of the printed text; `sep` (default to `\\` for ) specifies the separator between the values that are passed to the print() function.
