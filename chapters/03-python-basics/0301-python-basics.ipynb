### Basic Syntax

Python is designed to be readable and minimal, with syntax rules that emphasize clarity. There are basic rules that you have to follow when you write Python code:

- Case sensitive: Python is case sensitive with names; e.g., "Name" is different from "name". 
- Indentation: Python uses indentation (4 spaces) to define **blocks**, not {} braces.
- Comments:
  - \# The hash character or pound sign works as a single-line comment.
  - Multi-line comment: Python does not have multi-line comments, but you can use a multi-line string for the purpose. 

    ```code
    """
    Multi-line comment or docstring
    Used for documentation
    """
    ```
- Statement vs. Expression:
  - Statements are instructions that perform an action but do not produce a result value; e.g., x = 5 is an **assignment** statement, and to iterate through a sequence, we use a **for** statement or a **while** statement. 
  - Expressions are combinations of values, variables, and operators that the Python **interpreter** evaluates to produce a single value; e.g., len('hello') or "world".

## Variables

**Variables** are names that reference **objects** in memory. A couple of things to know about Python variables include:

- You create variables with the assignment operator (the equal sign, e.g., x = 2) and
- Can **reassign** them to different objects.
- Use descriptive, **lowercase** names with **underscores** for readability, and
- Avoid starting names with digits or symbols.

For naming conventions, see [PEP8](https://peps.python.org/pep-0008/) or [Google Python Style Guide](https://google.github.io/styleguide/pyguide.html).

### Python Object Defined

You would often here the saying that:
- Everything in Python is an **object**.

Two things we want to learn here: The definition of object in Python and how objects help us. 

**Definition**: Some other ideas to be considered about objects, however:  
- A variable doesn't "contain" a value; it **points** to an object in memory.
- Objects come from **classes**, which act like blueprints.
- Each object has:
  - **Identity** (where it lives in memory)
  - **Type** (what kind of thing it is)
  - **Value** (the data it holds)

To check the identy and type of an object, we can use the `id()` and `type()` method:

`````python
num = 100                                                            ### assign a value to a variable name/label
print("The unique identifier of this num object is: ", id(num))      ### identity of object
print("The type/class of this num object is: ", type(num))           ### data type of object
print("The value of this num object is: ", num)                      ### value of object
`````

## Operators

**Operators** enable you to perform common operations on variables. There are seven main operator types:

| No. | Operator type | Purpose | Examples |
| --- | --- | --- | --- |
| 1 | Assignment | Perform assignments and augmented assignments | `=`, `+=`, `:=`  |
| 2 | Arithmetic | Perform calculations (note `/` yields a floating-point number; avoid division by 0) | `+`, `-`, `*`, `/`, `%`, `//`, `**` |
| 3 | Comparison | Evaluate relationships | `==`, `!=`, `<`, `>`, `<=`, `>=` |
| 4 | Logical | Combine boolean expressions | `and`, `or`, `not` (e.g., `x == x and y > z`) |
| 5 | Identity | Compare object identity; equality compares values | `is`, `is not` (e.g., `a is b`) |
| 6 | Membership | Test membership in a collection | `in`, `not in` (e.g., `x in my_list`) |
| 7 | Bitwise | Bit-level operations | `&`, `\|`, `^`, `<<`, `>>` (e.g., `x & y`) |

### Operator Precedence

Python follows standard mathematical order of operations (PEMDAS: Parentheses, Exponents, Multiplication and Division from left to right, then Addition and Subtraction from left to right). For example:

`````python
print(2 + 3 * 5 + 5)
print(2 + 3 * (5 + 5))
`````

Strings are **sequences** of characters. You can access specific elements using **square bracket notation**. Python indexing starts at zero. Negative indexing in slicing starts with -1 from the end element.



`````python
s = 'hello'
print(s[0])
print(s[4])
print(s[-1])
`````

**Slice notation** allows you to grab parts of a string. Use a **colon** to specify the **start**, **stop**, and **step** indices in slicing. Note that the slicing operation is stop-**exclusive**.

`````python
s = 'abcdef'
print(s[0:])
print(s[:3])
print(s[3:6])
print(s[3:999])
print(s[0::2])
print(s[1:-1])
`````

### Lists

Lists are sequences of elements (similar to the array in many other languages) in **square brackets**, separated by _commas_. Lists can contain any data type, including other lists. 

`````python
my_list = ['a', 'b', 'c']

print(my_list)

my_list[0] = 'NEW'          ### update element (lists are mutable)
my_list.append('d')         ### using the append method
print(my_list)
`````

- Lists can be nested inside each other.
- Indexing and slicing work the same as in strings.
- You can access elements in nested lists by chaining square brackets.



`````python
lst = [1, 2, [3, 4]]

print(lst[0])
print(lst[1:3])             ### output a subset list
print(lst[2][1])            ### 4
`````


For deeper nesting, continue stacking brackets to access the desired element.



`````python
nest = [1, 2, 3, [4, 5, ['Target', 'Walmart']]]

print(nest[3][2][0])                  ### 'target'
`````

## Exceptions

According to [Python Documentation](https://docs.python.org/3/tutorial/errors.html), errors fall under two groups:

1. **Syntax errors**: Also known as **parsing errors**; Python detects the error when parsing/reading, namely, **before** executing your code.
2. **Exceptions**: Your statement or expression is syntactically correct but it causes an error when attempting to **execute**. Errors detected **during** execution are called exceptions.

Another way of looking at errors and exceptions is: An error is the problem and an exception is the mechanism Python uses to report that problem. When Python encounters a problem, it raises an exception and that exception produces an error message. In this sense, **all errors are exceptions**. In Python, there are many [built-in exceptions](https://docs.python.org/3/library/exceptions.html#BaseException) and they are instances of a class that derives from `BaseException`. Uses can define their own exceptions as well.

For a list of all built-in exceptions, see the [Python exception hierarchy](https://docs.python.org/3/library/exceptions.html#exception-hierarchy). The Python Standard Library has a list of 38 [concret exceptions](https://docs.python.org/3/library/exceptions.html#concrete-exceptions), but you will see some of those exceptions below more often than others: 

| No. | Error Type            | Meaning                       |
|-----| --------------------- | ----------------------------- |
|  1  | `SyntaxError`         | Code is written incorrectly   |
|  2  | `NameError`           | Variable/function not defined |
|  3  | `TypeError`           | Wrong data type used          |
|  4  | `ValueError`          | Type is right, value is wrong |
|  5  | `IndexError`          | List index doesn’t exist      |
|  6  | `KeyError`            | Dictionary key doesn’t exist  |
|  7  | `ZeroDivisionError`   | Dividing by zero              |

|  9  | `AttributeError`      | Wrong attribute for object    |
|  10 | `ModuleNotFoundError` | Import not installed/found    |
