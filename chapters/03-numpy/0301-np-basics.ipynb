# Arrays Basics

NumPy arrays (N-dimensional arrays, or **`ndarray`**) are a powerful extension of Python lists, offering compact storage and fast operations for numerical data in rows and columns.

While NumPy arrays can have any number of dimensions, the most common are:
- **Vectors**: **1-dimensional arrays** with a single axis
- **Matrices**: **2-dimensional arrays** with two axes (rows and columns)

This section covers core concepts for working with NumPy arrays:
- Array data structure
- Creating arrays
- Indexing and slicing
- Array attributes (ndim, dtype, shape) and functions like reshape()

## NumPy Data Structure

The primary data structure in NumPy is the N-dimensional array, or **`ndarray`**. NumPy's arrays are a list of lists in Python, but are more compact than Python lists. In essence, a Python list is an array of pointers to heterogeneous Python objects, while a NumPy array is an array of **uniform** values of the same type (e.g., all integers, all floats) and array elements are stored in one continuous block of memory, similar to how arrays work in C ({numref}`c-vs-python-list`). Python lists are more flexible, but Numpy arrays are smaller in file size, and access in reading and writing items is much faster {cite}`Martelli_2009`.

## Data Structure

The primary data structure in NumPy is the N-dimensional array, or **`ndarray`**. NumPy's arrays are a list of lists in Python, but are more compact than Python lists. In essence, a Python list is an array of pointers to heterogeneous Python objects, while a NumPy array is an array of **uniform** values of the same type (e.g., all integers, all floats) and array elements are stored in one continuous block of memory, similar to how arrays work in C ({numref}`c-vs-python-list`). Python lists are more flexible, but Numpy arrays are smaller in file size, and access in reading and writing items is much faster {cite}`Martelli_2009`, which is a huge advantage when dealing with large datasets.

## Array Attributes

Commonly used NumPy array attributes include:

- `ndim`: number of dimensions (rank)
- `shape`: tuple of lengths per axis (assignable to reshape if sizes match)
- `dtype`: data type of elements (e.g., float64, int32, complex128)

### dtype

The dtype property in NumPy is an attribute of a NumPy array (ndarray) that represents the data type of the elements contained within the array.

You can get the data type of the object in the array:

````python
arr = np.arange(25)
arr.dtype
````

````python
arr_string = np.array(['apple', 'banana', 'cherry'])

print(arr_string.dtype)
### U means UTF8, 6 is the longest string
````

````python
arr = np.array([1, 2, 3, 4], dtype='i4')

print(arr)
print(arr.dtype)
````

### ndim
The ndim attribute represents the number of dimensions (or axes) of the array.

````python
arr.ndim
````

````python
arr_25 = np.arange(25)
arr_55 = arr_25.reshape(5, 5)
arr_55.ndim
````

### shape

The shape attribute **returns a tuple** of integers at every index tells about the number of elements the corresponding dimension has.

````python
arr_2d = np.array([ [1, 2, 3], [4, 5, 6] ])
arr_2d.shape
````

````python
### create a 1-D array

arr_vec = np.arange(1, 10, 2)
arr_vec
````

## Creating NumPy Arrays

### From a Python List/Tuple

You can create an array by directly **casting** a list to an array. Let us create a list first:

````python
num_list = [1,2,3, 4, 5]
print(num_list)
print(type(num_list))
````

Now let's cast the list into a numpy array using numpy's `array()` method to create a **1-D** array.

````python
### before using numpy, import
### we usually import packages at the very beginning of a notebook

import numpy as np
````

````python
arr = np.array(num_list)       ### casting a list to a numpy array
arr

### now you will see a numpy array, which is array([ ele, ele, ele ]) from the list [1, 2, 3, 4, 5]
````

````python
### data type is different from list too: numpy.ndarray 

type(arr)                   
````

#### `dtype`

The **dtype** keyword allows you to explicitly set the data type of the resulting array:

````python
np.array([1, 2, 3, 4], dtype=np.float32)
````

### Using arange() Function

**`numpy.arange()`** is a NumPy function used to create arrays with evenly spaced values with a specified interval. It is similar to Python's built-in `range()` function, but returns a NumPy `ndarray` instead of a `range` object, which is more efficient for numerical operations.


````python
np.arange(5)
````

Using the range parameters: start, stop exclusive, stop:

````python
np.arange(2, 11, 2)
````

We can also place a Python **nested** list (list of lists) directly into the np.array function to create a NumPy array (note that this is a **2-D** array):

````python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])
print(arr)
````

An example of Python **3-D** list, which is a list of lists: 

````python
list_of_list = [
    [
        [1, 2, 3, 4], 
        [2, 2, 3, 4], 
        [3, 2, 3, 4]
    ], 
    [
        [4, 2, 3, 4], 
        [5, 2, 3, 4], 
        [6, 2, 3, 4]
    ]
]
list_of_list
````

Compare the result of the evaluation of the Python list above with the NumPy array created from the list:

````python
np.array(list_of_list)
````

### Using List Comprehension

Note that the syntax of list comprehension is an alternative to **for loops**. Instead of initializing an empty list variable and running a for loop to append elements, list comprehension is more concise. The syntax for list comprehension is:

```
[expression for item in iterable]
```

````python
np.array([i**2 for i in range(10)])
````

````python
np.array([range(i, i + 3) for i in [2, 4, 6]])
````

Or, list comprension with a condition as **filter**:
```
[expression for item in iterable if condition]
```


````python
evens = [x for x in range(10) if x % 2 == 0]
evens
````

### The reshape() Function 
The reshape() function changes the shape of a NumPy array without altering its underlying data. It **returns a new array** (or a view of the original array if possible) with the specified new shape.

````python
arr = np.linspace(0, 26, 27, dtype=int)
print(arr)
arr.reshape(3, 9)
print(arr)
````

## Indexing and Selection

Just like Python lists, square brackets `[ ]` are used for selecting the elements or groups of elements from an array:
1. Accessing individual elements of a list (**indexing**).
2. Selecting sub-sequences (**slicing** with parameters: start, stop exclusive, and step). Note that slicing returns a numpy array.
3. **Negative** indexing: begins with -1 from the end of the sequence.

````python
arr = np.arange(0,10)   ### creating sample array
arr
````

````python
print(arr)                  ### better looking of array output
````

````python

print("Shape:", arr.shape)  ### shape of the array
print("Dim  :", arr.ndim)   ### dimension of the array
````

````python
print(arr[1])               ### indexing 
print(arr[0:3])             ### slicing
print(arr[0:5:2])           ### slicing with step size
print(arr[-1])              ### negative indexing
print(arr[-3:-1])           ### slicing with negative indices
````

### Indexing/Slicing 2-D Aarrays (Matrices)

To access a 2-D array (matrices), use the syntax `array[row_index, column_index]`. The general format is **arr[row][col]** or **arr[row, col]**. 

````python
### creating a 2-d array

arr_2d = np.array(([5,10,15],[20,25,30],[35,40,45]))
arr_2d
````

````python
### 0-based indexing; 1st row
arr_2d[0]
````

````python
### 0-based indexing
arr_2d[1, 2]
````

````python
### 2D array slicing
### shape (2,2) from top right corner

arr_2d[:2, 1:]
````

### One More Example: 2-D Array (Matrix)

````python

arr_2d = np.array(
    (
        [5,10,15],
        [20,25,30],
        [35,40,45]
    )
)

#Show
arr_2d
````

````python
### indexing row
arr_2d[1]
````

````python
### Format is arr_2d[row][col] or arr_2d[row,col]

### Getting individual element value
arr_2d[1][0]     ### 2nd row, 1st column
````

````python
### getting individual element value
arr_2d[1,0]
````

````python
# 2D array slicing

### shape (2,2) from top right corner
arr_2d[:2,1:]
````

````python
### shape bottom row
arr_2d[2]
````

````python
### also shape bottom row with colon

arr_2d[2,:]
````
