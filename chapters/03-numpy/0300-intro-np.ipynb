# NumPy Arrays

NumPy, or Numerical Python, is the foundation for Pythonic data science, machine learning, and scientific computing libraries such as **pandas**, **scikit-learn**, and **TensorFlow**. Learning NumPy means understanding the basics of how these other tools work. NumPy (or Numpy) is basically a linear algebra library for Python. As a numerical library for Python, NumPy is important for data science because almost all libraries in the Python data science ecosystem rely on it as a building block.
