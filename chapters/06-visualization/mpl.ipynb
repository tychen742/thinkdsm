## The Data

```python
### sample data

x = np.linspace(0, 10, 100)
y = np.sin(x)

x_len = len(x)
x_size = x.size    
x_max = max(x)     ### function
x_min = x.min()    ### method
x_std = x.std()
x_sum = x.sum()
# x_count = x.count(0)    ### list: count the value passed in the object

x_info = [                ### make a list of tuples
    ('x_len', x_len),
    ('x_size', x_size),
    ('x_max', x_max),
    ('x_min', x_min),
    ('x_std', x_std),
    ('x_sum', x_sum)
]

### loop through a list of tuples
for name, val in x_info:     ### this is called tuple unpacking
    # print(name, val)
    print(f"{name:10} ==> {val}")

```

### Customizing Plot
- label
    - The label argument is what you assign to a plotted line or object.
    - It doesn’t display automatically; instead, it’s used by the legend.
    - The plt.legend() call tells Matplotlib to draw a legend using all labeled artists (lines, patches, etc.).
    - If you don’t call plt.legend(), the labels won’t be shown.
- title
- xlabel
- ylabel
- legend
    - plt.legend(loc='upper right', title='Functions', fontsize=10)
- grid
- show

```python
### plot it 
plt.plot(x, y, label='sin(x)', color='grey', linestyle='--')

# add labels and title
plt.title('Basic Sine Wave')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.legend()
plt.grid(True)
# plt.show()      ### == Jupyter %matplotlib inline
```

## Q03. Bar Chart

```python
### sample data
categories = [ 'A' , 'B' , 'c' , 'D' ]
values = [25, 40, 55, 30] 

### create a bar plot 
plt.figure(figsize=(5, 3))                   ### figsize
plt.bar(categories, values, color='teal')    ### categories, values, color (color is in the Other Parameter section)
plt.title('Categorical Data Comparison') 
plt.xlabel('Category')                       ### string literal
plt.ylabel('Value')                          ### string literal
plt. show()                                  ### %mpl inline
```

## Q04. Scatter Plots

Scatter plots are used to visualize the relationship between two continuous variables. Each point represents a single observation. 

```python
### sample data 
x = np.random.rand(50) * 10         ### rand: random float 0-1 exclusive
y = np.random.rand(50) * 10 
colors = np.random.rand(50)         ### colors for each point; interesting
sizes = 100 * np.random.rand(50)    ### randomizing sizes of each point

### creating a scatter plot 
plt.figure(figsize=(5, 3))          ### (5, 3) tutple; personal favorite
                                    ### MPL DEFUALT is plt.figure(figsize=(6.4, 4.8))
                                    ### Units are in inches
                                    ### Default DPI (dots per inch) is usually 100
                                    ### typical Matplotlib window looks like a 640×480 image — 
                                    ### the same size as an old VGA screen
plt.scatter(x, y, c=colors, s=sizes, alpha=0.7, cmap='viridis') 
plt.title('Random Scatter Plot') 
plt.xlabel('X-axis') 
plt.ylabel('Y-axis') 
plt.colorbar(label='Color intensity') 
plt.show() 
```


## Histograms

Histograms are used to visualize the distribution of a single numerical variable. They show how many data points fall into different ranges (bins). 

```python
### sample data 

np.random.seed(42)
data = np.random.randn(1000)      ### rand, randn, randint
# Create a histogram 
# plt.figure(figsize=(8, 6)) 
plt.figure(figsize=(5, 3)) 
plt.hist(data, bins=30, color='purple', alpha=0.75, edgecolor='white' )

plt.title('Distribution of Random Data') 
plt.xlabel('Value') 
plt.ylabel('Frequency')
```
