# Pandas DataFrames

DataFrames are the core data structure in Pandas, modeled after similar structures in the **R** programming language. A DataFrame is a **two-dimensional, labeled data structure** that can be thought of as:

- A **collection of Series objects** combined into a single table, all sharing the same index
- A **spreadsheet or SQL table** with rows and columns
- A **dictionary of Series**, where each column is a Series with a shared index

**Key Features of DataFrames**

- **Labeled axes**: Both rows (index) and columns have labels, making data access intuitive and explicit
- **Heterogeneous data**: Different columns can contain different data types (integers, floats, strings, dates, etc.)
- **Size-mutable**: You can insert and delete columns
- **Powerful operations**: Built-in methods for filtering, grouping, merging, reshaping, and statistical analysis
- **Alignment** of data based on labels
- **Missing data** handling (see [0404-missing-data.ipynb](0404-missing-data.ipynb) for details)
- **Rich I/O tools** for reading/writing various formats (CSV, Excel, SQL, JSON, etc.)

**Why DataFrames?**

While NumPy arrays are excellent for numerical computation, DataFrames offer several advantages:

- **Column names** instead of just numeric indices
- **Mixed data types** in different columns
- Accessing and selecting data
- Modifying DataFrames (adding/removing columns and rows)
- Sorting and filtering
- Working with indices, including MultiIndex

- **`.iat[]`** - position-based, for single value access

Use these when you only need one value, as they're optimized for speed.

````python
# Using .at[] for fast label-based access to a single value
df.at['Alice', 'Salary']
````

````python
# Using .iat[] for fast position-based access
# Get value at row 0, column 0 (Alice's Salary)
df.iat[0, 0]
````

````python
# Performance comparison: .at[] is faster than .loc[] for single values
# .loc[] works, but .at[] is optimized for this use case
print("Using .loc[]:", df.loc['Bob', 'Rating'])
print("Using .at[]:", df.at['Bob', 'Rating'])
````

#### Multiple Conditions

**Common Pitfall**: Use `&` and `|`, NOT `and` or `or`. Python's `and`/`or` don't work with pandas boolean indexing.

For two or more conditions, use `|` (OR) and `&` (AND) with parentheses.

````python
### Combine conditions with & (and) or | (or)
### Each condition must be in parentheses

df[(df['Salary']>70) & (df['Years'] > 2)]
````

#### Alternative: Using .query() Method

The `.query()` method offers a more readable way to filter DataFrames, especially with multiple conditions.

````python
# Same result as above, but more readable
df.query('Salary > 70 and Years > 2')
````

````python
# .query() also supports variables using @ symbol
min_salary = 70
df.query('Salary > @min_salary')
````

## Applying Functions with `.apply()`

The `.apply()` method is a powerful tool that allows you to apply a function along an axis of the DataFrame (rows or columns). This is useful for transformations, calculations, and custom operations.

The syntax for `.apply()` is:

```python
DataFrame.apply(func, axis=0, raw=False, result_type=None, args=(), **kwargs)
```

**Parameters:**
- `func`: Function to apply to each column or row
- `axis`: {0 or 'index', 1 or 'columns'}, default 0
  - 0 or 'index': apply function to each column
  - 1 or 'columns': apply function to each row
- `raw`: bool, default False. If False, passes each row/column as a Series to the function. If True, passes ndarray
- `result_type`: {'expand', 'reduce', 'broadcast', None}, default None
- `args`: tuple, positional arguments to pass to func
- `**kwargs`: additional keyword arguments to pass to func

**Returns:** Series or DataFrame

**Performance Note:** For simple operations like arithmetic or comparisons, vectorized operations are usually faster than `.apply()`. Use `.apply()` when you need to apply a complex function that can't be vectorized.

```python
# Vectorized (faster): df['A'] * 2
# Using apply (slower): df['A'].apply(lambda x: x * 2)
```

````python
df
````

````python
# Apply a function to each column (axis=0, default)
print(df.apply(np.sum))
print(type(df.apply(np.sum)))
````

````python
# Apply a function to each row (axis=1)
df.apply(np.sum, axis=1)
````

````python
# Apply a custom function (double each value)
df.apply(lambda x: x * 2)
````

### Multi-Index and Index Hierarchy

**Hierarchical Indexing** (or **MultiIndex**) allows multiple levels of row or column labels in a DataFrame. This is very useful for working with grouped or multidimensional data, such as time series across multiple categories or nested groupings.

**When to use MultiIndex:**

- Working with panel data (e.g., multiple time series)
- Hierarchical groupings (e.g., Country > State > City)
- Multi-dimensional data that doesn't fit well in a flat table

Let's create a quick example using `pd.MultiIndex.from_tuples()`.

### Creating multi-level index

````python
# Create index levels for MultiIndex
outside = ['G1','G1','G1','G2','G2','G2']
inside = [1,2,3,1,2,3]

# In Python, zip() combines two or more lists element-by-element into tuples
# Example:
#   list1 = [1, 2, 3, 4]
#   list2 = ['a', 'b']
#   list(zip(list1, list2)) => [(1, 'a'), (2, 'b')]

hier_index = list(zip(outside,inside))
hier_index = pd.MultiIndex.from_tuples(hier_index)
````

````python
hier_index     # A MultiIndex created from tuples
````

### Creating MultiIndex DataFrame

Pass the hier_index to index parameter; also add columns.

````python
df = pd.DataFrame(np.random.randn(6,2),index=hier_index,columns=['A','B'])
df
````

Now let's show how to index this! For index hierarchy we use **df.loc[ ]**, if this was on the columns axis, you would just use normal bracket notation df[]. Calling one level of the index returns the sub-dataframe:

````python
df.loc['G1']
````

## Performance Tips and Best Practices

Understanding how Pandas handles data internally can help you write more efficient code.

### Copy vs View

When you slice or select data from a DataFrame, Pandas sometimes returns a **view** (reference to original data) and sometimes a **copy** (new data). This affects both performance and behavior.

````python
# Create a sample DataFrame
perf_df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
print("Original DataFrame:")
print(perf_df)
````

````python
# Selecting a column returns a view (changes affect original)
col_a = perf_df['A']
print("\nColumn A:", col_a)
print("Is it a view? Changes to col_a affect perf_df")
````

````python
# Use .copy() to create an independent copy
col_a_copy = perf_df['A'].copy()
col_a_copy[0] = 999

print("\nAfter modifying col_a_copy:")
print("Original DataFrame A column:", perf_df['A'][0])  # Unchanged
print("Copy:", col_a_copy[0])  # Changed
````

**Best Practice:** Use `.copy()` when you want to ensure your modifications don't affect the original DataFrame.

### Memory Usage

Use `.memory_usage()` to check how much memory your DataFrame is using.

````python
# Check memory usage in bytes
perf_df.memory_usage()
````

````python
# Get total memory in KB or MB
total_bytes = perf_df.memory_usage(deep=True).sum()
print(f"Total memory: {total_bytes} bytes")
print(f"Total memory: {total_bytes / 1024:.2f} KB")
````

### Vectorization vs Iteration

**Always prefer vectorized operations over loops** when working with DataFrames. Vectorized operations are much faster.

````python
# SLOW: Iterating through rows (avoid this!)
# for index, row in df.iterrows():
#     df.at[index, 'new_col'] = row['A'] * 2

# FAST: Vectorized operation
perf_df['A_doubled'] = perf_df['A'] * 2
print(perf_df)
````

**Key Takeaways:**

1. Use `.copy()` when you need an independent copy of data
2. Check `.memory_usage()` for large datasets
3. Prefer vectorized operations over loops
4. Use `.at[]` and `.iat[]` for single-value access instead of `.loc[]` and `.iloc[]`
5. Chain operations carefully to avoid unintended side effects
