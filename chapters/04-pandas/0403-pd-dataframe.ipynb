# Pandas DataFrames

DataFrames are the core data structure in Pandas, modeled after similar structures in the **R** programming language. A DataFrame is a **two-dimensional, labeled data structure** that can be thought of as:

- A **collection of Series objects** combined into a single table, all sharing the same index
- A **spreadsheet or SQL table** with rows and columns
- A **dictionary of Series**, where each column is a Series with a shared index

**Key Features of DataFrames**

- **Labeled axes**: Both rows (index) and columns have labels, making data access intuitive and explicit
- **Heterogeneous data**: Different columns can contain different data types (integers, floats, strings, dates, etc.)
- **Size-mutable**: You can insert and delete columns
- **Powerful operations**: Built-in methods for filtering, grouping, merging, reshaping, and statistical analysis
- **Alignment** of data based on labels
- **Missing data** handling with NaN
- **Rich I/O tools** for reading/writing various formats (CSV, Excel, SQL, JSON, etc.)

**Why DataFrames?**

While NumPy arrays are excellent for numerical computation, DataFrames offer several advantages:

- **Column names** instead of just numeric indices
- **Mixed data types** in different columns
- Handling missing data
- Accessing and selecting data
- Modifying DataFrames (adding/removing columns and rows)
- Sorting and filtering
- Working with indices, including MultiIndex

## Handling Missing Data

Missing data is extremely common in real-world datasets. Pandas represents missing values with `NaN` (Not a Number) and provides powerful tools to detect, remove, or fill missing data.

````python
# Create a DataFrame with missing values for demonstration
import numpy as np

df_missing = pd.DataFrame({
    'A': [1, 2, np.nan, 4],
    'B': [5, np.nan, np.nan, 8],
    'C': [9, 10, 11, 12]
})
df_missing
````

### Detecting Missing Data

````python
# Check for missing values (returns boolean DataFrame)
df_missing.isna()
````

````python
# Alternative: isnull() is an alias for isna()
df_missing.isnull()
````

````python
# Count missing values per column
df_missing.isna().sum()
````

### Dropping Missing Data

````python
# Drop any row with at least one NaN value
df_missing.dropna()
````

````python
# Drop columns with any NaN values
df_missing.dropna(axis=1)
````

````python
# Drop rows where ALL values are NaN (more conservative)
df_missing.dropna(how='all')
````

````python
# Drop rows with fewer than 2 non-NaN values
df_missing.dropna(thresh=2)
````

### Filling Missing Data

````python
# Fill all NaN values with a specific value
df_missing.fillna(0)
````

````python
# Fill with mean of each column
df_missing.fillna(df_missing.mean())
````

````python
# Fill with different values for different columns
df_missing.fillna({'A': 0, 'B': df_missing['B'].mean()})
````

````python
# Forward fill (use previous valid value)
df_missing.fillna(method='ffill')
````

````python
# Backward fill (use next valid value)
df_missing.fillna(method='bfill')
````

## Applying Functions with .apply()

The `.apply()` method is a powerful tool that allows you to apply a function along an axis of the DataFrame (rows or columns). This is useful for transformations, calculations, and custom operations.

### Syntax

```python
DataFrame.apply(func, axis=0, raw=False, result_type=None, args=(), **kwargs)
```

**Parameters:**
- `func`: Function to apply to each column or row
- `axis`: {0 or 'index', 1 or 'columns'}, default 0
  - 0 or 'index': apply function to each column
  - 1 or 'columns': apply function to each row
- `raw`: bool, default False. If False, passes each row/column as a Series to the function. If True, passes ndarray
- `result_type`: {'expand', 'reduce', 'broadcast', None}, default None
- `args`: tuple, positional arguments to pass to func
- `**kwargs`: additional keyword arguments to pass to func

**Returns:** Series or DataFrame

````python
df
````

````python
# Apply a function to each column (axis=0, default)
print(df.apply(np.sum))
print(type(df.apply(np.sum)))
````

````python
# Apply a function to each row (axis=1)
df.apply(np.sum, axis=1)
````

````python
# Apply a custom function (double each value)
df.apply(lambda x: x * 2)
````

````python
# Apply to a single column
df['W'].apply(lambda x: x ** 2)
````

## Multi-Index and Index Hierarchy (Advanced Topic)

**Hierarchical Indexing** (or **MultiIndex**) allows multiple levels of row or column labels in a DataFrame. This is very useful for working with grouped or multidimensional data, such as time series across multiple categories or nested groupings.

**When to use MultiIndex:**

- Working with panel data (e.g., multiple time series)Let's create a quick example using `pd.MultiIndex.from_tuples()`.

- Hierarchical groupings (e.g., Country > State > City)
- Multi-dimensional data that doesn't fit well in a flat table

### Creating multi-level index

````python
# Create index levels for MultiIndex
outside = ['G1','G1','G1','G2','G2','G2']
inside = [1,2,3,1,2,3]

# In Python, zip() combines two or more lists element-by-element into tuples
# Example:
#   list1 = [1, 2, 3, 4]
#   list2 = ['a', 'b']
#   list(zip(list1, list2)) => [(1, 'a'), (2, 'b')]

hier_index = list(zip(outside,inside))
hier_index = pd.MultiIndex.from_tuples(hier_index)
````

````python
hier_index     # A MultiIndex created from tuples
````

### Creating MultiIndex DataFrame

Pass the hier_index to index parameter; also add columns.

````python

df = pd.DataFrame(np.random.randn(6,2),index=hier_index,columns=['A','B'])
df
````

Now let's show how to index this! For index hierarchy we use **df.loc[ ]**, if this was on the columns axis, you would just use normal bracket notation df[]. Calling one level of the index returns the sub-dataframe:

````python
df.loc['G1']
````

````python
df.loc['G1'].loc[1]
````

````python
df.loc['G2'].loc[2]['B']
````
