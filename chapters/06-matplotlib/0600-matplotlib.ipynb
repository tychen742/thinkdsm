# Matplotlib

Matplotlib gives Python users direct control over figures, axes, labels, legends, styles, and common chart types. This chapter moves from quick `pyplot` commands to the object-oriented plotting style used for clearer, more customizable visualizations.

## Learning Goals

By the end of this chapter, you should be able to:

1. Create basic Matplotlib plots with `pyplot`.
2. Explain the difference between a figure and an axes.
3. Use `plt.subplots()` to build object-oriented plots.
4. Control labels, titles, legends, ranges, colors, line styles, and figure size.
5. Choose common Matplotlib plot types for business and data analysis tasks.
