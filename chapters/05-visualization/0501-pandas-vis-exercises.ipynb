# Pandas Visualization – Exercises

Work through the exercises below. Each question cell is interactive. Solution cells are hidden by default (click to reveal).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('ggplot')
np.random.seed(42)

df1 = pd.DataFrame({'A': np.random.randn(1000), 'B': np.random.randn(1000), 'C': np.random.rand(1000)})
dates = pd.date_range('2020-01-01', periods=100, freq='D')
df2 = pd.DataFrame({'a': np.random.rand(100), 'b': np.random.rand(100), 'c': np.random.rand(100)}, index=dates)
df = pd.DataFrame(np.random.randn(1000, 2), columns=['a', 'b'])
```

## Exercise 1 – Histogram Fundamentals

**Goal:** Visualize the distribution of `A` in `df1`.

**Requirements:**
1. Plot a histogram for `df1['A']` with exactly **50 bins**.
2. Set `figsize=(8, 4)` and `alpha=0.7`.
3. Add title `"Distribution of A"`.

**Stretch:** Add vertical dashed lines for the mean and median of `A`.

```python
### Your code here
```

```python
### Solution
ax = df1['A'].plot.hist(bins=50, figsize=(8, 4), alpha=0.7)
ax.set_title('Distribution of A')
ax.axvline(df1['A'].mean(), color='red', linestyle='--', label='mean')
ax.axvline(df1['A'].median(), color='blue', linestyle='--', label='median')
ax.legend()
plt.show()
```

## Exercise 2 – Line Plot Styling

**Goal:** Show trend behavior for one feature.

**Requirements:**
1. Plot `df1['B']` as a line plot.
2. Use `figsize=(8, 4)` and `lw=2`.
3. Set x-label `"Index"` and y-label `"B values"`.

**Stretch:** Overlay a rolling mean (`window=10`) on the same axes.

```python
### Your code here
```

```python
### Solution
ax = df1.plot.line(y='B', figsize=(8, 4), lw=2)
ax.set_xlabel('Index')
ax.set_ylabel('B values')
df1['B'].rolling(10).mean().plot(ax=ax, lw=2, color='red', label='rolling mean')
ax.legend()
plt.show()
```

## Exercise 5 – Size-Encoded Scatter

**Goal:** Add a third variable using marker size.

**Requirements:**
1. Plot `A` vs `B` from `df1`.
2. Use point size `abs(df1['C']) * 60`.
3. Set `alpha=0.5`.

**Stretch:** Compare size encoding to color encoding and note trade-offs.

```python
### Your code here
```

```python
### Solution
ax = df1.plot.scatter(x='A', y='B', s=df1['C'].abs() * 60, alpha=0.5, figsize=(7, 5))
ax.set_title('A vs B (size = |C| x 60)')
plt.show()
```

## Exercise 8 – KDE for a Single Variable

**Goal:** Plot a smooth estimate of one distribution.

**Requirements:**
1. Plot KDE for `df2['a']`.
2. Set `figsize=(8, 4)`.
3. Add title `"KDE of column a"`.

**Stretch:** Overlay histogram + KDE in one figure for `df2['a']`.

```python
### Your code here
```

```python
### Solution
ax = df2['a'].plot.kde(figsize=(8, 4))
ax.set_title('KDE of column a')
df2['a'].plot.hist(bins=20, alpha=0.4, density=True, ax=ax)
plt.show()
```
