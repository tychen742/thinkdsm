### CREATE: Define Tables

```python
import sqlite3

# Connect to a database file (creates it if it doesn't exist)
conn = sqlite3.connect('mydatabase.db')

# Create a cursor to execute SQL commands
cursor = conn.cursor()

# Execute SQL commands using the cursor
cursor.execute("""CREATE TABLE employees (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    department TEXT
)""")

# Save changes
conn.commit()

# Close the connection
conn.close()
```

## How SQLite Works

SQLite stores data in a single binary file (typically with `.db` or `.sqlite` extension). The file contains:
- **Table definitions** (structure and constraints)
- **Index information** (for fast lookups)
- **Actual data** (all your records)

This is different from server-based databases like MySQL, which require a separate server process running and listening for connections.

## Why SQLite for This Lab?

### Advantages of SQLite:

1. **No Server Required**: It's embedded directly in Python without needing a running database server
2. **Simple Setup**: Works with just a `.db` file; no complex configuration
3. **Python Integration**: Python includes SQLite support built-in through the `sqlite3` module
4. **Full SQL Support**: Implements standard SQL, so skills transfer to MySQL, PostgreSQL, and others
5. **Zero Configuration**: Just create a database and start using it

### Trade-offs:

- **Limited Concurrency**: Not ideal for many simultaneous users
- **Single Machine**: Can't be accessed over a network (unlike MySQL/PostgreSQL)
- **Smaller Scale**: Better for smaller datasets (still millions of records possible)

## Introduction to SQLite

**SQLite** is a lightweight, file-based relational database system that's perfect for learning SQL and building applications without the complexity of setting up a separate database server. Unlike MySQL or PostgreSQL, SQLite stores everything in a single file on your computer, making it ideal for:

- **Learning SQL** without complex installation
- **Small to medium applications** and local projects
- **Mobile applications** (used by iOS, Android)
- **Quick prototyping** and testing
- **Educational purposes** (like this course!)
