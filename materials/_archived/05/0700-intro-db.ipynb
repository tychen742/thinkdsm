## Common Relational Database Systems

- **SQLite**: Lightweight, file-based (ideal for learning and small applications)
- **MySQL**: Popular open-source database used in web applications
- **PostgreSQL**: Advanced open-source with powerful features
- **SQL Server**: Enterprise-level database from Microsoft
- **Oracle**: Large-scale enterprise database system

## Advantages of Relational Databases

1. **Data Integrity**: Constraints ensure data consistency and validity
2. **Efficient Queries**: SQL enables fast, powerful queries on large datasets
3. **Scalability**: Can handle millions of records efficiently
4. **Reduced Redundancy**: Relationships eliminate duplicate data storage
5. **Flexibility**: Easy to modify structure and add new tables

## Non-Relational (NoSQL) Databases

While relational databases are powerful for structured data, **non-relational databases** (often called **NoSQL** databases) have emerged to handle different types of data and use cases. These databases prioritize flexibility, scalability, and performance over the strict structure of tables and relationships.

**Common Types of NoSQL Databases:**
- **Document Databases** (MongoDB, CouchDB): Store data as flexible JSON-like documents instead of rigid rows
- **Key-Value Stores** (Redis, Memcached): Store simple key-value pairs for fast retrieval, ideal for caching
- **Column-Family Stores** (HBase, Cassandra): Optimize for column-based queries and massive scale
- **Graph Databases** (Neo4j): Designed for data with complex relationships and network patterns
- **Search Engines** (Elasticsearch): Optimized for full-text search and analytics

**JSON in NoSQL:** Most modern NoSQL databases use **JSON** (JavaScript Object Notation) as their native data format. JSON allows flexible, hierarchical data structures without requiring a predefined schema:

```json
{
  "employeeId": 1,
  "firstName": "John",
  "lastName": "Smith",
  "department": {
    "id": 10,
    "name": "Engineering"
  },
  "skills": ["Python", "SQL", "JavaScript"],
  "contact": {
    "email": "john.smith@company.com",
    "phone": "555-1234"
  }
}
```

Unlike relational databases that flatten this data into multiple tables, document databases store the entire structure as a single document. This flexibility makes NoSQL ideal for evolving data structures, but sacrifices some of the consistency guarantees that relational databases provide.

**When to Choose NoSQL:**
- Handling unstructured or semi-structured data (images, text, logs)
- Rapid prototyping when data structure may change
- Massive scalability across distributed systems
- Real-time analytics and big data processing
- Complex relationship queries (graph databases)

**When to Stick with Relational:**
- Highly structured, consistent data
- Complex queries across multiple tables
- Strong need for data integrity and ACID guarantees
- Traditional business applications (finance, HR, inventory)
- When your data naturally fits into tables with clear relationships

**In This Course:** We focus on relational databases because they provide the foundational concepts for understanding all database systems. The principles of keys, relationships, and normalization appear in nearly every modern database, making SQL and relational thinking essential knowledge for any data professional.

## SQL: The Language of Relational Databases

**SQL** (Structured Query Language) is the standard language for interacting with relational databases. It allows you to:
- **SELECT**: Retrieve data from tables
- **INSERT**: Add new data
- **UPDATE**: Modify existing data
- **DELETE**: Remove data
- **CREATE**: Define new tables and relationships

## Key Concepts

### Tables, Rows, and Columns

Relational databases organize data into **tables** (also called relations). Each table represents an entity or concept:
- **Columns** (or fields) represent attributes or properties of the entity
- **Rows** (or records) represent individual instances of that entity
- **Primary Keys** uniquely identify each row in a table

### Example: Employee Table

| EmployeeID | FirstName | LastName | Department | Salary |
|---|---|---|---|---|
| 1 | John | Smith | Engineering | 85000 |
| 2 | Jane | Doe | Marketing | 72000 |
| 3 | Bob | Johnson | Engineering | 88000 |

### Relationships Between Tables

The power of relational databases comes from their ability to establish **relationships** between tables:
- A single table might reference data from another table using **foreign keys**
- This avoids data duplication and keeps data consistent
- Common relationship types: one-to-one, one-to-many, many-to-many

### Example: Departments and Employees

**Departments Table**
| DeptID | DeptName |
|---|---|
| 1 | Engineering |
| 2 | Marketing |

**Employees Table** (with DeptID as foreign key)
| EmployeeID | FirstName | DeptID |
|---|---|---|
| 1 | John | 1 |
| 2 | Jane | 2 |
| 3 | Bob | 1 |

## RDB Review

### Primary Keys

A **primary key** is a column (or set of columns) that uniquely identifies each row in a table. Key characteristics:
- **Uniqueness**: No two rows can have the same primary key value
- **Non-null**: Every row must have a primary key value (cannot be empty)
- **Immutability**: Should not change over time
- **Single per table**: Each table has exactly one primary key

**Example:**
```
EmployeeID (Primary Key) | FirstName | LastName
         1               | John      | Smith
         2               | Jane      | Doe
         3               | Bob       | Johnson
```

In this table, `EmployeeID` is the primary key because it uniquely identifies each employee.

### Foreign Keys

A **foreign key** is a column that references the primary key of another table. It establishes relationships between tables:
- **Referential Integrity**: Ensures data consistency by enforcing relationships
- **Links Tables**: Connects rows in one table to rows in another
- **Allows Relationships**: Enables one-to-many and many-to-many relationships

**Example:**
```
Employees Table:
EmployeeID (PK) | FirstName | DepartmentID (FK)
         1      | John      |        10
         2      | Jane      |        20
         3      | Bob       |        10

Departments Table:
DepartmentID (PK) | DepartmentName
         10       | Engineering
         20       | Marketing
```

Here, `DepartmentID` in the Employees table is a foreign key referencing `DepartmentID` in the Departments table.

### Candidate Keys

A **candidate key** is any column (or set of columns) that could serve as a primary key:
- Must have the uniqueness property
- A table can have multiple candidate keys
- One candidate key is chosen as the primary key
- Others are alternate keys

**Example:** In an Employees table, both `EmployeeID` and `Email` could be candidate keys (both are unique), but `EmployeeID` is chosen as the primary key.

### Composite Keys

A **composite key** (or compound key) is a primary key made up of multiple columns:
- Used when a single column cannot uniquely identify a row
- All columns together must be unique
- Common in junction tables for many-to-many relationships

**Example: Student Course Enrollments**
```
StudentID | CourseID | Semester
    1001  |   CS101  |   Fall2023
    1001  |   CS102  |   Fall2023
    1002  |   CS101  |   Fall2023
```

Here, the combination of `StudentID + CourseID + Semester` forms the primary key (composite key).

### Constraints

**Constraints** are rules that ensure data quality and consistency:

- **NOT NULL**: Column must always have a value
- **UNIQUE**: All values in the column must be different
- **PRIMARY KEY**: Uniquely identifies each row (NOT NULL + UNIQUE)
- **FOREIGN KEY**: References primary key in another table
- **CHECK**: Ensures values meet a specific condition
- **DEFAULT**: Provides a default value if none is specified

**Example:**
```sql
CREATE TABLE Employees (
    EmployeeID INTEGER PRIMARY KEY,
    Email TEXT NOT NULL UNIQUE,
    FirstName TEXT NOT NULL,
    Salary REAL CHECK (Salary > 0),
    Status TEXT DEFAULT 'Active'
);
```

### Normalization

**Normalization** is the process of organizing data to minimize redundancy and improve integrity. Key principles:

- **Reduce Duplicates**: Avoid storing the same information multiple times
- **Maintain Relationships**: Use foreign keys instead of duplicate data
- **First Normal Form (1NF)**: Each cell contains only atomic (indivisible) values
- **Second Normal Form (2NF)**: All non-key columns depend on the entire primary key
- **Third Normal Form (3NF)**: Non-key columns depend only on the primary key

**Example: Unnormalized vs. Normalized**

Unnormalized (redundant):
```
EmployeeID | Name | Department | Manager
    1      | John | Engineering| Sarah
    2      | Jane | Marketing  | Tom
    1      | John | Engineering| Sarah  (duplicate!)
```

Normalized (efficient):
```
Employees Table:
EmployeeID | Name | DepartmentID
    1      | John |      10
    2      | Jane |      20

Departments Table:
DepartmentID | Department | Manager
     10      | Engineering| Sarah
     20      | Marketing  | Tom
```

## Introduction

A **relational database** is a structured system for storing and organizing data using tables with rows and columns. It's one of the most widely-used data storage systems in practice because it provides a robust, efficient, and scalable way to manage large amounts of data.
